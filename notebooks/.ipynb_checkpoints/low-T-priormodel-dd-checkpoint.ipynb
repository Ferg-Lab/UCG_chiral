## Prepare for saving potentials at target temperature

```python
valid_mask_ll = ucg_comRDF['t1']['rdf'] > 1.0E-40
valid_mask_dd = ucg_comRDF['t2']['rdf'] > 1.0E-40
valid_mask_ld = ucg_comRDF['t12']['rdf'] > 1.0E-40 

#ELIMINATE ZERO RDF BINS
xll = ucg_comRDF['t1']['bin_centers'] #[valid_mask_ll]
xdd = ucg_comRDF['t2']['bin_centers'] #[valid_mask_dd]
xld = ucg_comRDF['t12']['bin_centers'] #[valid_mask_ld]

xll_valid = ucg_comRDF['t1']['bin_centers'][valid_mask_ll]
xdd_valid = ucg_comRDF['t2']['bin_centers'][valid_mask_dd]
xld_valid = ucg_comRDF['t12']['bin_centers'][valid_mask_ld]

ull_pot_valid = ull_pot[valid_mask_ll]
udd_pot_valid  = udd_pot[valid_mask_dd]
uld_pot_valid  = uld_pot[valid_mask_ld]

print(f'Starting bins for spline fits in prior model. xll:{xll[0]} xll:{xdd[0]} xld:{xld[0]}')
```

```python
from scipy.interpolate import make_interp_spline

udd_pot[0] = 330 # 750 #330
udd_pot[1] = 250 #540 #250
udd_pot[2] = 140 #410 #140
udd_pot[3] = 80 # 330 #120 #80
udd_pot[4] = 30 #250 #90 #60
udd_pot[5] = 24 #140 # 80 #50


# # CASE 1:
# udd_pot = ull_pot.copy()
# # CASE 2:
ull_pot = udd_pot.copy()

# for i in np.arange(-1,-10,-1):
#     print(i)
#     ull_pot[i]=ull_pot[-11]

uld_pot[0] = 750
uld_pot[1] = 540
uld_pot[2] = 410
uld_pot[3] = 330
uld_pot[4] = 250
uld_pot[5] = 140
# uld_pot[6] = 80
# uld_pot[7] = 60
# uld_pot[8] = 50
# uld_pot[9] = 45
# uld_pot[10] = 41





# uld_pot[(xld<1)] = ull_pot[(xll<1)]

# uld_pot[-1] = 0
# uld_pot[-2] = 0
# uld_pot[-3] = 0
# uld_pot[-4] = 0
# uld_pot[-5] = 0
# uld_pot[-6] = 0
# uld_pot[-7] = 0
# uld_pot[-8] = 0
# uld_pot[-9] = 0
# uld_pot[-10] = 0

# udd_pot[0] = 680
# udd_pot[1] = 410
# udd_pot[2] = 330
# udd_pot[3] = 250
# # udd_pot[4] = 140
# # udd_pot[5] = 80
# # udd_pot[6] = 60
# # udd_pot[7] = 50


ull_pot_adjusted = ull_pot.copy() 
ull_pot_adjusted-=ull_pot_adjusted[-1]
n = 3
cut_max=xll[-1]
# ll_spline = make_interp_spline(xll[(xll>0.8) & (xll<cut_max)], ull_pot_adjusted[(xll>.8) & (xll<cut_max)], k=n)
ll_spline = make_interp_spline(xll[(xll>0.) & (xll<cut_max)], ull_pot_adjusted[(xll>0.) & (xll<cut_max)], k=n)


## BEST VALUES FOR 2.9 -- low-t (CASE 1 -- SEE ABOVE)
## BEST VALUES for 3.0 -- low-t-2 (CASE 2 -- SEE ABOVE)
uld_pot_adjusted = uld_pot.copy()*1
# uld_pot_adjusted[(xll>5.)] = uld_pot_adjusted[(xll<5.)][-1]
uld_pot_adjusted-=uld_pot_adjusted[-1]
n = 1
cut_max=xld[-1]
ld_spline = make_interp_spline(xld[(xld>0.) & (xld<cut_max)], uld_pot_adjusted[(xld>0.) & (xld<cut_max)], k=n)
# ld_spline = make_interp_spline(xld, uld_pot_adjusted, k=n)


udd_pot_adjusted = udd_pot.copy()
udd_pot_adjusted-=udd_pot_adjusted[-1]
n = 3
cut_max=xdd[-1]
# dd_spline = make_interp_spline(xdd[(xdd>0.7) & (xdd<cut_max)], udd_pot_adjusted[(xdd>0.7) & (xdd<cut_max)], k=n)
dd_spline = make_interp_spline(xdd[(xdd>0.) & (xdd<cut_max)], udd_pot_adjusted[(xdd>0.) & (xdd<cut_max)], k=n)



xll_fine = np.linspace(0.1,7, 50)
xld_fine = np.linspace(0.1,7, 50)
xdd_fine = np.linspace(0.1,7, 50)

ll_combined = ll_spline(xll_fine)
ll_combined[(xll_fine>6)] = 0.0
ld_combined = ld_spline(xld_fine) 
ld_combined[(xld_fine>6.)] = 0.0
dd_combined = dd_spline(xdd_fine) 
dd_combined[(xdd_fine>6.)] = 0.0
```

## Save prior potentials to tables

```python
results_path = '/project2/andrewferguson/sivadasetty/doe/ucg/notebooks/chiral/manuscript/simulations/all-t/'

ref_t = kt_reduced

# explore_temperature_list = [4.6] 
temperature = 2.9
explore_temperature_list = [temperature]
# explore_temperature_list = [2.7]


for new_t in explore_temperature_list:
    for gamma in [0]:
    #for gamma in [-2, -1, -0.5, 0.5, 1, 2]:
        

        target_folder = results_path + 't_' + str(new_t) + '/gamma_' + str(gamma) + '/'
        #target_folder = results_path + 'master_4.6/'
        
        scaled_t = (new_t/ref_t)**gamma

        print(scaled_t)

        scaled_ull_pot = scaled_t*ll_combined 
        scaled_udd_pot = scaled_t*dd_combined 
        scaled_uld_pot = scaled_t*ld_combined 

        ull_force = -np.gradient(scaled_ull_pot,xll_fine[1]-xll_fine[0])
        scaled_udd_pot = scaled_ull_pot
        udd_force = ull_force # -np.gradient(scaled_udd_pot,0.2)#xdd_fine[1]-xdd_fine[0])
        uld_force = -np.gradient(scaled_uld_pot,xld_fine[1]-xld_fine[0])

        # Open the file in write mode
        with open(target_folder + '/prior_model/Pair_L-L.table', 'w') as file:
            file.write('# Table Pair_L-L: id, r, potential, force' + '\n')  # Add a newline character between rows
            file.write('\n')
            file.write('L-L\n')
            file.write('N ' + str(len(ull_force)) + ' R ' + str(xll_fine.min()*1) + ' ' + str(xll_fine.max()*1.0))
            file.write('\n\n')
            
            for i in range(ull_force.shape[0]):
        
                file.write(str(i+1) + ' ' + str(xll_fine[i]*1) + ' ' + str(scaled_ull_pot[i]) + ' ' + str(ull_force[i]))
                file.write('\n')
                
        # Open the file in write mode
        with open(target_folder + '/prior_model/Pair_L-D.table', 'w') as file:
            file.write('# Table Pair_L-D: id, r, potential, force' + '\n')  # Add a newline character between rows
            file.write('\n')
            file.write('L-D\n')
            file.write('N ' + str(len(uld_force)) + ' R ' + str(xld_fine.min()*1) + ' ' + str(xld_fine.max()*1.0))
            file.write('\n\n')
            
            for i in range(uld_force.shape[0]):
                file.write(str(i+1) + ' ' + str(xld_fine[i]*1) + ' ' + str(scaled_uld_pot[i]) + ' ' + str(uld_force[i]))
                file.write('\n')
                
        # Open the file in write mode
        with open(target_folder + '/prior_model/Pair_D-D.table', 'w') as file:
            file.write('# Table Pair_D-D: id, r, potential, force' + '\n')  # Add a newline character between rows
            file.write('\n')
            file.write('D-D\n')
            file.write('N ' + str(len(udd_force)) + ' R ' + str(xdd_fine.min()*1) + ' ' + str(xdd_fine.max()*1.0))
            file.write('\n\n')
            
            for i in range(udd_force.shape[0]):
                file.write(str(i+1) + ' ' + str(xdd_fine[i]*1) + ' ' + str(scaled_udd_pot[i]) + ' ' + str(udd_force[i]))
                file.write('\n')        
        
                
        
```
