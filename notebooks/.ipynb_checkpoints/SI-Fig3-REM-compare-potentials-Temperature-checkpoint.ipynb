```python
from scipy.interpolate import BSpline

degree = 3
k = degree

xmin, xmax = 0.1, 6.0
interior = np.array([0.5, 0.9, 1.2, 1.5, 1.8, 2.1, 2.4, 2.7, 3.0, 3.5, 4.0, 4.5, 5.0], float)

t = np.concatenate([np.full(k+1, xmin), interior, np.full(k+1, xmax)])
n_basis = len(t) - k - 1

nlike_left =   6 #2 #2 #4
nlike_right = 6
nunlike_left = 7 #2 #2 #6
nunlike_right = 6 #6 #5 #6 #5

flike_left =  2 
flike_right = 6
funlike_left = 2 
funlike_right = 6 
```

```python
##### 
fig = plt.figure(figsize=(10, 8))
plt.subplots_adjust(left=0.1, right=0.9, bottom=0.1, top=0.9, wspace=0.35, hspace=0.3)
grid = GridSpec(2, 2)
n = maxiter - startiter # number of colors
colors = plt.cm.rainbow(np.linspace(0, 1, n))  # or use 'plasma', 'inferno', 'tab10', etc.

for k, iter in enumerate(range(startiter,maxiter)):
    
    saved_pot_ll = pd.read_csv(results_path + '/iter_'+str(iter)+'/Pair_L-L.table', skiprows=4, 
                               sep='\s+', names=['r', 'u', 'f'])
    saved_pot_dd = pd.read_csv(results_path + '/iter_'+str(iter)+'/Pair_D-D.table', skiprows=4, 
                               sep='\s+', names=['r', 'u', 'f'])
    saved_pot_ld = pd.read_csv(results_path + '/iter_'+str(iter)+'/Pair_L-D.table', skiprows=4, 
                               sep='\s+', names=['r', 'u', 'f'])
    
    # outsaved_pot_ll = pd.read_csv(results_path + '/init_potential'+'/outPair_L-L.table', skiprows=5, 
    #                            sep='\s+', names=['r', 'u', 'f'])
    # outsaved_pot_dd = pd.read_csv(results_path + '/init_potential'+'/outPair_D-D.table', skiprows=5, 
    #                            sep='\s+', names=['r', 'u', 'f'])
    # outsaved_pot_ld = pd.read_csv(results_path + '/init_potential'+'/outPair_L-D.table', skiprows=5, 
    #                            sep='\s+', names=['r', 'u', 'f'])

    coeff_like, coeff_unlike, _ = load_coeffs(results_path + '/iter_'+str(iter)+'/coeff.pkl')
    

    xj = [saved_pot_ll.r, saved_pot_ld.r]
    yj = [saved_pot_ll.u, saved_pot_ld.u]

    # oxj = [outsaved_pot_ll.r, outsaved_pot_dd.r, outsaved_pot_ld.r]
    # oyj = [outsaved_pot_ll.u, outsaved_pot_dd.u, outsaved_pot_ld.u]
    
    for j in range(2):
        ax1 = plt.subplot(grid[j, 0])
        # ax1.grid()
        
        # ax1.scatter(oxj[j], oyj[j], color=colors[iter], s=5)
        # ax1.plot(oxj[j], oyj[j], color=colors[iter], lw=2)

        ax1.scatter(xj[j], yj[j], color=colors[k], s=5)
        ax1.plot(xj[j], yj[j], color=colors[k], lw=2)

        if k == 0:

            if j == 0:
                # -----------------------------
                # Build ALL unscaled basis values B[m, :]
                # -----------------------------
                B = np.zeros((n_basis, xj[j].size))
                for m in range(n_basis):
                    cm = np.zeros(n_basis)
                    cm[m] = 3.0 #coeff_like[m]
                    Bm = BSpline(t, cm, degree, extrapolate=False)
                    B[m, :] = Bm(xj[j])
                
                # Dominant basis index at each x (UNSCALED)
                #dominant = np.argmax(B, axis=0)
                
                basiscolors = plt.cm.rainbow(np.linspace(0, 1, n_basis-nlike_left-nlike_right))
                # basiscolors = ['blue', 'cyan', 'green', 'orange', 'red']
                # (A) Plot ALL basis functions (unscaled), colored by index
                mm = 0
                for m in range(n_basis):
                    if m < flike_left or m >= n_basis-flike_right:
                        ax1.fill_between(xj[j], B[m, :], color='black', linewidth=1.4, alpha=0.25)
                    elif m < nlike_left and m >= flike_left:
                        ax1.fill_between(xj[j], B[m, :], color='gray', linewidth=1.4, alpha=0.25)
                    elif m >= n_basis-nlike_right and m < n_basis-flike_right:
                        ax1.fill_between(xj[j], B[m, :], color='gray', linewidth=1.4, alpha=0.25)
                    else:
                        ax1.fill_between(xj[j], B[m, :], color=basiscolors[mm], linewidth=1.4, alpha=0.25)
                        mm += 1

            if j == 1:
                # -----------------------------
                # Build ALL unscaled basis values B[m, :]
                # -----------------------------
                B = np.zeros((n_basis, xj[j].size))
                for m in range(n_basis):
                    cm = np.zeros(n_basis)
                    cm[m] = 3.0 #coeff_unlike[m] #1.0
                    Bm = BSpline(t, cm, degree, extrapolate=False)
                    B[m, :] = Bm(xj[j])

                
                basiscolors = plt.cm.rainbow(np.linspace(0, 1, n_basis-nunlike_left-nunlike_right))
                # basiscolors = ['blue', 'cyan', 'green', 'orange', 'red']
                
                # (A) Plot ALL basis functions (unscaled), colored by index
                mm = 0
                for m in range(n_basis):
                    if m < funlike_left or m >= n_basis-funlike_right:
                        ax1.fill_between(xj[j], B[m, :], color='black', linewidth=1.4, alpha=0.25)
                    elif m < nunlike_left and m >= funlike_left:
                        ax1.fill_between(xj[j], B[m, :], color='gray', linewidth=1.4, alpha=0.25)
                    elif m >= n_basis-nunlike_right and m < n_basis-funlike_right:
                        ax1.fill_between(xj[j], B[m, :], color='gray', linewidth=1.4, alpha=0.25)
                    else:
                        ax1.fill_between(xj[j], B[m, :], color=basiscolors[mm], linewidth=1.4, alpha=0.25)
                        mm += 1

        
        
        ax1.set_xlabel(r"$r$")
        ax1.set_ylabel("U($r$)")
    
        ax1.set_ylim(-25,5)
        # ax1.set_xlim(0., 2)
        
        ax1.xaxis.set_major_locator(MultipleLocator(1))
        ax1.yaxis.set_major_locator(MultipleLocator(2))
        ax1.xaxis.set_minor_locator(AutoMinorLocator(n=5))
        ax1.yaxis.set_minor_locator(AutoMinorLocator(n=5))
        ## ax1.set_ylim(0.5,2)
        ## ax1.set_ylim(0.5,2)

        # ax1.axvline(x=.7)
        # ax1.axvline(x=1.5)
        # ax1.axvline(x=1.25)
    
        # ax1.grid()

    xj = [saved_pot_ll.r, saved_pot_ld.r]
    yj = [saved_pot_ll.f, saved_pot_ld.f]
    for j in range(2):
        ax2 = plt.subplot(grid[j, 1])
        # ax2.grid()
        ax2.scatter(xj[j], yj[j], color=colors[k], s=5)
        ax2.plot(xj[j], yj[j], color=colors[k], lw=2)
        
        # ax2.grid()
        ax2.set_ylim(-30,5)#00)
        # ax2.set_xlim(1,2)
        
        ax2.set_xlabel(r"$r$")
        ax2.set_ylabel("F($r$)")
    
        # ax2.axvline(x=.7)
        # ax2.axvline(x=.1)
        # ax2.axvline(x=1.5)
        
        ax2.xaxis.set_major_locator(MultipleLocator(1))
        ax2.yaxis.set_major_locator(MultipleLocator(2)) #0.5))
        ax2.xaxis.set_minor_locator(AutoMinorLocator(n=5))
        ax2.yaxis.set_minor_locator(AutoMinorLocator(n=5))

for j in range(2):
    ax1 = plt.subplot(grid[j, 0])
    ax1.grid()
    ax2 = plt.subplot(grid[j, 1])
    ax2.grid()
        
```
