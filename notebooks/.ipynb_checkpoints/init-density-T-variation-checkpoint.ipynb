```python
import numpy as np
from scipy.signal import find_peaks, savgol_filter  # or gaussian_filter1d

def histogram_peaks(counts, bin_centers, smooth=True, window=7, poly=2,
                    prominence=0.05, distance=1, width=None):
    """
    Return peak x-positions (bin centers) from a histogram.
    - counts: array of histogram heights (len N)
    - bin_edges: array of length N+1
    - smooth: apply Savitzky–Golay smoothing before peak finding
    - window: odd window length (in bins) for smoothing
    - poly: polynomial order for Savitzky–Golay
    - prominence: min peak prominence as a fraction of max(counts) if 0<val<1; absolute if >=1
    - distance: min separation between peaks in bins
    - width: optional min peak width in bins (float or None)
    """
    counts = np.asarray(counts)
    if smooth and len(counts) >= window:
        y = savgol_filter(counts, window_length=window, polyorder=poly, mode="interp")
    else:
        y = counts

    # Convert relative prominence to absolute if given as fraction
    prom = prominence * y.max() if 0 < prominence < 1 else prominence

    idx, props = find_peaks(y, prominence=prom, distance=distance, width=width)
    x_peaks = bin_centers[idx]
    return x_peaks, idx, props  # x positions, indices into counts, and peak properties

```
