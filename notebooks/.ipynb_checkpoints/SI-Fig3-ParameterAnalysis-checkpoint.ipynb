## Read REM potentials and check

```python
targetiters = {}
for density in [0.11, 0.09, 0.07]:
    tpath = {}
    if density == 0.11:
        for temperature in [3.8, 4.2, 4.5, 4.6, 5.0]:
            if temperature == 5.0:
                tpath[temperature] = 772 #370 #772
            elif temperature == 4.6:
                tpath[temperature] = 591 #364 # 591
            elif temperature == 4.5:
                tpath[temperature] = 649 #282 #649 #282
            elif temperature == 4.2:
                tpath[temperature] = 435
            elif temperature == 3.8:
                tpath[temperature] = 239
    elif density == 0.09:
        for temperature in [3.8]:
            tpath[temperature] = 519
    elif density == 0.07:
        for temperature in [3.8]:
            tpath[temperature] = 450
            
    targetiters[density] = tpath


        

main_path = '/project2/andrewferguson/sivadasetty/doe/ucg/notebooks/chiral/manuscript/simulations/rem/high-t-repeat-nneigh12-NR-Hess-7-EXTENDED'
master_path = main_path + '/master/'
results_path = {}
pot_results_path = {}

sys.path.append(master_path)
from write_potentials import load_coeffs, load_loss, load_models


for density in targetiters.keys():
    tpath = {}
    ptpath = {}
    for temperature in targetiters[density].keys():

        if temperature in [4.5, 4.6, 5.0]:
            chiral_lag = 100
        else:
            chiral_lag = 50

        tpath[temperature] = main_path + '/t_'+str(temperature)+'_chiral_lag_'+str(chiral_lag)+'_nsteps_50000/iter_' + str(targetiters[density][temperature]) + '/'
        ptpath[temperature] = main_path + '/t_'+str(temperature)+'_chiral_lag_'+str(chiral_lag)+'_nsteps_50000/iter_' + str(targetiters[density][temperature]) + '/'

    results_path[density] = tpath 
    pot_results_path[density] = ptpath
        
```

```python
from scipy.interpolate import BSpline

degree = 3
k = degree

xmin, xmax = 0.1, 6.0
interior = np.array([0.5, 0.9, 1.2, 1.5, 1.8, 2.1, 2.4, 2.7, 3.0, 3.5, 4.0, 4.5, 5.0], float)

t = np.concatenate([np.full(k+1, xmin), interior, np.full(k+1, xmax)])
print(t)
n_basis = len(t) - k - 1

nlike_left =  6 #7 #6 
nlike_right = 6 
nunlike_left = 8 #9 #8
nunlike_right = 6 

# ---- range dictionary ----
coeff_support = {
    i: [float(t[i]), float(t[i + k + 1])]
    for i in range(n_basis)
}

# # print nicely
# for i, r in coeff_support.items():
#     print(f"c_{i}: [{r[0]:.2f}, {r[1]:.2f})")
```

## Convergence of gradients and parameters

```python

loss = {}
sigmoid = {}
clike = {}
cunlike = {}
saved_pot_ll = {}
saved_pot_dd = {}
saved_pot_ld = {}

for density in targetiters.keys():
    
    loss_t = {}
    sigmoid_t = {}
    clike_t = {}
    cunlike_t = {}
    saved_pot_ll_t = {}
    saved_pot_dd_t = {}
    saved_pot_ld_t = {}

    for temperature in targetiters[density].keys():
        
        if os.path.exists(results_path[density][temperature]+'/loss.pkl'):
            loss_t[temperature] = load_loss(results_path[density][temperature] +'/loss.pkl')
            
        if os.path.exists(results_path[density][temperature] + '/coeff.pkl'):    
            clike_t[temperature], cunlike_t[temperature], sigmoid_t[temperature] = load_coeffs( results_path[density][temperature] + '/coeff.pkl')

        saved_pot_ll_t[temperature] = pd.read_csv(pot_results_path[density][temperature] + '/Pair_L-L.table', skiprows=4, 
                                                  sep='\s+', names=['r', 'u', 'f'])
        saved_pot_dd_t[temperature] = pd.read_csv(pot_results_path[density][temperature] + '/Pair_D-D.table', skiprows=4, 
                                                  sep='\s+', names=['r', 'u', 'f'])
        saved_pot_ld_t[temperature] = pd.read_csv(pot_results_path[density][temperature] + '/Pair_L-D.table', skiprows=4, 
                                                  sep='\s+', names=['r', 'u', 'f'])
    
    loss[density] = loss_t
    sigmoid[density] = sigmoid_t
    clike[density] = clike_t
    cunlike[density] = cunlike_t
    saved_pot_ll[density] = saved_pot_ll_t
    saved_pot_dd[density] = saved_pot_dd_t
    saved_pot_ld[density] = saved_pot_ld_t
    
    
```
