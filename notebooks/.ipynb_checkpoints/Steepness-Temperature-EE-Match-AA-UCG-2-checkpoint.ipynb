```python
################################################################################################
## SOME HELPFUL FUNCTIONS
################################################################################################

import pandas as pd

def read_lammps_dump_xyz(file_path):

    with open(file_path, 'r') as f:
        lines = f.readlines()
    
    timesteps = {}
    current_timestep = None
    header_line = None
    data_lines = []

    for i, line in enumerate(lines):
        
        line = line.strip()
        
        if line.startswith("ITEM: TIMESTEP"):
            if current_timestep is not None and header_line and data_lines:
                df = pd.DataFrame(data_lines, columns=header_line)
                timesteps[current_timestep] = df
                data_lines = []  # Reset data for the new timestep
            
            current_timestep = int(lines[i + 1].strip())
            
            if current_timestep > 20774000:
                break

            
            header_line = None # reset
        
        elif line.startswith("ITEM: ATOMS"):
            header_line = line.split()[2:] 

        elif line and header_line is not None and not line.startswith("ITEM:"):  
            data_lines.append(list(map(float, line.split())))
            
    if current_timestep is not None and header_line and data_lines:
        df = pd.DataFrame(data_lines, columns=header_line)
        timesteps[current_timestep] = df
    
    return timesteps


def write_lammps_dump_from_dict(timesteps_dict, box_bounds, out_trajfile, selected_timesteps):

    with open(out_trajfile, 'w') as f:
        for ts in sorted(selected_timesteps):
            if ts not in timesteps_dict:
                continue  # Skip timesteps not in the dictionary

            df = timesteps_dict[ts]
            num_atoms = len(df)

            f.write("ITEM: TIMESTEP\n")
            f.write(f"{ts}\n")
            f.write("ITEM: NUMBER OF ATOMS\n")
            f.write(f"{num_atoms}\n")
            f.write("ITEM: BOX BOUNDS pp pp pp\n")
            f.write("0.0 " + str(box_bounds[0]) + "\n")
            f.write("0.0 " + str(box_bounds[1]) + "\n")
            f.write("0.0 " + str(box_bounds[2]) + "\n")
            f.write("ITEM: ATOMS " + ' '.join(df.columns) + "\n")

            for row in df.itertuples(index=False):
                row = list(row)
                row_out = [f"{int(row[0])}", f"{int(row[1])}"] + [f"{val:.6f}" for val in row[2:]]
                f.write(' '.join(row_out) + "\n")


def compute_rdf(traj, box_bounds, bins, start, stop, min_bound, max_bound, normalization=True, openmscgnorm=False):

    timesteps = sorted(traj.keys())
    n_frames = len(timesteps)
    if stop == -1:
        stop = n_frames

    #min_bound, max_bound = 0, box_bounds[0]/2.0
    volume = box_bounds[0]*box_bounds[1]*box_bounds[2]
    
    rdf_list_type_1 = []
    rdf_list_type_2 = []
    rdf_list = []
    rdf_list_all = []
    
    for timestep in range(start, stop): # timesteps:

        # select_atoms_type_1 = np.array(traj[timestep][ (traj[timestep]['c_compute'] == -1.0) ][['x', 'y', 'z']])
        # select_atoms_type_2 = np.array(traj[timestep][ (traj[timestep]['c_compute'] == 1.0) ][['x', 'y', 'z']])
        select_atoms_type_1 = np.array(traj[timestep][ (traj[timestep]['c_compute'] < 0.0) ][['x', 'y', 'z']])
        select_atoms_type_2 = np.array(traj[timestep][ (traj[timestep]['c_compute'] >= 0.0) ][['x', 'y', 'z']])
        select_atoms_all = np.array(traj[timestep][['x', 'y', 'z']])
        

        # compute distances
        pairdists_type_1 = distances.self_distance_array(select_atoms_type_1, box=box_bounds)
        if not openmscgnorm:
            num_density_type_1 = ((len(select_atoms_type_1)-1)*len(select_atoms_type_1))/volume
        else:
            num_density_type_1 = 1/volume

        pairdists_type_2 = distances.self_distance_array(select_atoms_type_2, box=box_bounds)
        if not openmscgnorm:
            num_density_type_2 = ((len(select_atoms_type_2)-1)*len(select_atoms_type_2))/volume
        else:
            num_density_type_2 = 1/volume
        
        pairdists = distances.distance_array(select_atoms_type_1, select_atoms_type_2, box=box_bounds)
        if not openmscgnorm:
            num_density = (len(select_atoms_type_1)*len(select_atoms_type_2))/volume
        else:
            num_density = 1/volume

        pairdists_all = distances.self_distance_array(select_atoms_all, box=box_bounds)
        #pairdists_all = np.array([x for x in pairdists_all if x <= max_bound])
        if not openmscgnorm:
            num_density_all = ((len(select_atoms_all)-1)*len(select_atoms_all))/volume
        else:
            num_density_all = 1/volume

        hist_type_1, bin_edges_type_1 = np.histogram(pairdists_type_1, bins=bins, range=[min_bound, max_bound])
        hist_type_2, bin_edges_type_2 = np.histogram(pairdists_type_2, bins=bins, range=[min_bound, max_bound])
        hist, bin_edges = np.histogram(pairdists, bins=bins, range=[min_bound, max_bound])
        hist_all, bin_edges_all = np.histogram(pairdists_all, bins=bins, range=[min_bound, max_bound])
        
        bin_centers_type_1 = (bin_edges_type_1[1:]+bin_edges_type_1[:-1])/2.0
        bin_centers_type_2 = (bin_edges_type_2[1:]+bin_edges_type_2[:-1])/2.0
        bin_centers = (bin_edges[1:]+bin_edges[:-1])/2.0
        bin_centers_all = (bin_edges_all[1:]+bin_edges_all[:-1])/2.0
        
        vols_type_1 = np.power(bin_edges_type_1, 3)
        vols_type_2 = np.power(bin_edges_type_2, 3)
        vols = np.power(bin_edges, 3)
        vols_all = np.power(bin_edges_all, 3)
        
        norm_type_1 = 4 / 3 * np.pi * np.diff(vols_type_1)
        norm_type_2 = 4 / 3 * np.pi * np.diff(vols_type_2)
        norm = 4 / 3 * np.pi * np.diff(vols)
        norm_all = 4 / 3 * np.pi * np.diff(vols_all)

        if timestep %100 == 0:
            print(f'{timestep} of {stop} completed. Num each type : {len(select_atoms_type_1)} {len(select_atoms_type_2)} {len(select_atoms_all)}')
        
        norm_type_1 *= num_density_type_1
        norm_type_2 *= num_density_type_2
        norm *= num_density
        norm_all *= num_density_all

        if normalization:
            rdf_type_1 = hist_type_1/(norm_type_1)
            rdf_type_2 = hist_type_2/(norm_type_2)
            rdf = hist/(norm)
            rdf_all = hist_all/(norm_all)
        else:
            rdf_type_1 = hist_type_1
            rdf_type_2 = hist_type_2
            rdf = hist
            rdf_all = hist_all
            

        rdf_list_type_1.append(rdf_type_1)
        rdf_list_type_2.append(rdf_type_2)
        rdf_list.append(rdf)
        rdf_list_all.append(rdf_all)

    rdf_avg_type_1 = np.mean(rdf_list_type_1, axis=0)
    rdf_avg_type_2 = np.mean(rdf_list_type_2, axis=0)
    rdf_avg = np.mean(rdf_list, axis=0)
    rdf_avg_all = np.mean(rdf_list_all, axis=0)
    
    # Note bin centers are automatically from last frame
    rdf_data_t_1 = {'bin_centers': bin_centers_type_1, 'rdf': rdf_avg_type_1}
    rdf_data_t_2 = {'bin_centers': bin_centers_type_2, 'rdf': rdf_avg_type_2}
    rdf_data = {'bin_centers': bin_centers, 'rdf': rdf_avg}
    rdf_data_all = {'bin_centers': bin_centers_all, 'rdf': rdf_avg_all}
    
    
    return {'t1': rdf_data_t_1, 't2': rdf_data_t_2, 't12': rdf_data, 'all': rdf_data_all}
    


def target_compute_rdf(traj, box_bounds, bins, start, stop, min_bound, max_bound, normalization=True):

    timesteps = sorted(traj.keys())
    n_frames = len(timesteps)
    if stop == -1:
        stop = n_frames

    #min_bound, max_bound = 0, box_bounds[0]/2.0
    volume = box_bounds[0]*box_bounds[1]*box_bounds[2]
    
    rdf_list_type_1 = []
    rdf_list_type_2 = []
    rdf_list = []
    rdf_list_all = []
    
    #for timestep in range(start, stop): # timesteps:
    for ts, timestep in enumerate(traj):

        if ts == stop:
            break

        select_atoms_type_1 = np.array(traj[timestep][ (traj[timestep]['c_compute_chirality'] == 1.0) ][['x', 'y', 'z']])
        select_atoms_type_2 = np.array(traj[timestep][ (traj[timestep]['c_compute_chirality'] == 0.0) ][['x', 'y', 'z']])
        select_atoms_all = np.array(traj[timestep][['x', 'y', 'z']])
        
        
        # compute distances
        pairdists_type_1 = distances.self_distance_array(select_atoms_type_1, box=box_bounds)
        num_density_type_1 = ((len(select_atoms_type_1)-1)*len(select_atoms_type_1))/volume

        pairdists_type_2 = distances.self_distance_array(select_atoms_type_2, box=box_bounds)
        num_density_type_2 = ((len(select_atoms_type_2)-1)*len(select_atoms_type_2))/volume
        
        pairdists = distances.distance_array(select_atoms_type_1, select_atoms_type_2, box=box_bounds)
        num_density = (len(select_atoms_type_1)*len(select_atoms_type_2))/volume

        pairdists_all = distances.self_distance_array(select_atoms_all, box=box_bounds)
        #pairdists_all = np.array([x for x in pairdists_all if x <= max_bound])
        num_density_all = ((len(select_atoms_all)-1)*len(select_atoms_all))/volume

        hist_type_1, bin_edges_type_1 = np.histogram(pairdists_type_1, bins=bins, range=[min_bound, max_bound])
        hist_type_2, bin_edges_type_2 = np.histogram(pairdists_type_2, bins=bins, range=[min_bound, max_bound])
        hist, bin_edges = np.histogram(pairdists, bins=bins, range=[min_bound, max_bound])
        hist_all, bin_edges_all = np.histogram(pairdists_all, bins=bins, range=[min_bound, max_bound])
        
        bin_centers_type_1 = (bin_edges_type_1[1:]+bin_edges_type_1[:-1])/2.0
        bin_centers_type_2 = (bin_edges_type_2[1:]+bin_edges_type_2[:-1])/2.0
        bin_centers = (bin_edges[1:]+bin_edges[:-1])/2.0
        bin_centers_all = (bin_edges_all[1:]+bin_edges_all[:-1])/2.0
        
        vols_type_1 = np.power(bin_edges_type_1, 3)
        vols_type_2 = np.power(bin_edges_type_2, 3)
        vols = np.power(bin_edges, 3)
        vols_all = np.power(bin_edges_all, 3)
        
        norm_type_1 = 4 / 3 * np.pi * np.diff(vols_type_1)
        norm_type_2 = 4 / 3 * np.pi * np.diff(vols_type_2)
        norm = 4 / 3 * np.pi * np.diff(vols)
        norm_all = 4 / 3 * np.pi * np.diff(vols_all)

        if ts % 10 == 0:
            print(f'{ts} of {stop} completed. Num each type : {len(select_atoms_type_1)} {len(select_atoms_type_2)} {len(select_atoms_all)}')
        
        norm_type_1 *= num_density_type_1
        norm_type_2 *= num_density_type_2
        norm *= num_density
        norm_all *= num_density_all

        if normalization:
            rdf_type_1 = hist_type_1/(norm_type_1)
            rdf_type_2 = hist_type_2/(norm_type_2)
            rdf = hist/(norm)
            rdf_all = hist_all/(norm_all)
        else:
            rdf_type_1 = hist_type_1
            rdf_type_2 = hist_type_2
            rdf = hist
            rdf_all = hist_all
            

        rdf_list_type_1.append(rdf_type_1)
        rdf_list_type_2.append(rdf_type_2)
        rdf_list.append(rdf)
        rdf_list_all.append(rdf_all)

    rdf_avg_type_1 = np.mean(rdf_list_type_1, axis=0)
    rdf_avg_type_2 = np.mean(rdf_list_type_2, axis=0)
    rdf_avg = np.mean(rdf_list, axis=0)
    rdf_avg_all = np.mean(rdf_list_all, axis=0)
    
    # Note bin centers are automatically from last frame
    rdf_data_t_1 = {'bin_centers': bin_centers_type_1, 'rdf': rdf_avg_type_1}
    rdf_data_t_2 = {'bin_centers': bin_centers_type_2, 'rdf': rdf_avg_type_2}
    rdf_data = {'bin_centers': bin_centers, 'rdf': rdf_avg}
    rdf_data_all = {'bin_centers': bin_centers_all, 'rdf': rdf_avg_all}
    
    
    return {'t1': rdf_data_t_1, 't2': rdf_data_t_2, 't12': rdf_data, 'all': rdf_data_all}






def target_compute_averageChirality(traj, start, stop):

    timesteps = sorted(traj.keys())
    n_frames = len(timesteps)
    if stop == -1:
        stop = n_frames

    num_l = {}
    num_d = {}
    for ts, timestep in enumerate(traj):

        if ts == stop:
            break

        select_atoms_type_1 = np.array(traj[timestep][ (traj[timestep]['c_compute_chirality'] == 1.0) ][['x', 'y', 'z']])
        select_atoms_type_2 = np.array(traj[timestep][ (traj[timestep]['c_compute_chirality'] == 0.0) ][['x', 'y', 'z']])
        select_atoms_all = np.array(traj[timestep][['x', 'y', 'z']])
        
        num_type_1 = len(select_atoms_type_1)
        num_type_2 = len(select_atoms_type_2)
        num_all = len(select_atoms_all)

        num_l[int(ts)] = num_type_1
        num_d[int(ts)] = num_type_2

        if ts % 10 == 0:
            print(f'{ts} of {stop} completed. Num each type : {num_type_1} {num_type_2} {num_all}')
        

    return {'l':num_l, 'd':num_d}



def aa_compute_averageChirality(traj, start, stop):

    timesteps = sorted(traj.keys())
    n_frames = len(timesteps)
    if stop == -1:
        stop = n_frames

    num_l = {}
    num_d = {}
    for ts, timestep in enumerate(traj):

        if ts == stop:
            break

        select_atoms_type_1 = np.array(traj[timestep][ (traj[timestep]['c_compute'] < 0.0) ][['x', 'y', 'z']])
        select_atoms_type_2 = np.array(traj[timestep][ (traj[timestep]['c_compute'] >= 0.0) ][['x', 'y', 'z']])
        select_atoms_all = np.array(traj[timestep][['x', 'y', 'z']])
        
        num_type_1 = len(select_atoms_type_1)
        num_type_2 = len(select_atoms_type_2)
        num_all = len(select_atoms_all)

        num_l[int(ts)] = num_type_1
        num_d[int(ts)] = num_type_2

        if ts % 10 == 0:
            print(f'{ts} of {stop} completed. Num each type : {num_type_1} {num_type_2} {num_all}')
        

    return {'l':num_l, 'd':num_d}
    


```

## UCG EE analysis

```python
################################################################################################

ucg_ee = {}
ucg_ee_average = {}
ucg_ee_std = {}
for s in steepness_list:
    for temperature in temperature_list:

        steepness = np.round(s,3)
        print(f'Steepness:{steepness} temperature:{temperature}')
        case= 'ucg_a'+str(steepness)+'_t'+str(temperature)
        results_path = '/project2/andrewferguson/sivadasetty/doe/ucg/notebooks/chiral/steepness-temperature-explore/'+case+'/'
        target_count_results_file = results_path + '/chiral_counts_ucg.pkl'
        if os.path.exists(target_count_results_file):
            with open(target_count_results_file, 'rb') as file:
                ucg_chiral_counts[case] = pickle.load(file)

        ee_num = np.array(list(ucg_chiral_counts[case]['l'].values())) - np.array(list(ucg_chiral_counts[case]['d'].values()))
        ee_denom = np.array(list(ucg_chiral_counts[case]['l'].values())) + np.array(list(ucg_chiral_counts[case]['d'].values()))
        ucg_ee[case] = ee_num/ee_denom
        ucg_ee_average[case] = np.mean(ee_num/ee_denom)
        ucg_ee_std[case] = np.std(ee_num/ee_denom)
        
        
```

## Match AA EE distribution from EE vs T and UCG EE distribution from EE vs S

```python
# === KL Divergence Function ===
def kl_divergence(p, q):
    return np.sum(p * np.log(p / q))

# === Main Comparison Function ===
def compute_kl_matches(dict1, dict2, inner_key='ee_symmetry_hist'):
    """
    dict1: reference AA (temperature keys as outer keys)
    dict2: target UCG (steepness keys as outer keys)
    inner_key: key to access the PMF in each dictionary
    """
    results = []

    for k1, v1 in dict1.items():
        p = np.clip(v1[inner_key], 1e-10, None)  # Ensure numerical stability

        best_k2, best_kl = None, float('inf')

        for k2, v2 in dict2.items():
            q = np.clip(v2[inner_key], 1e-10, None)
            
            if len(p) != len(q):
                print('p and q lenghts not matching')
                continue  # only compare matching shapes

            kl = kl_divergence(p, q)
            results.append((k1, k2, kl))

            if kl < best_kl:
                best_k2, best_kl = k2, kl

    df_all = pd.DataFrame(results, columns=["T", "S", "KL_Divergence"])
    df_best = df_all.sort_values(by=["T", "KL_Divergence"]).groupby("T").first().reset_index()

    return df_all, df_best
```

```python
full_table, best_matches = compute_kl_matches(aa_ee_symmetry_hist_temperature, ucg_ee_symmetry_hist_temperature)

print("All KL Divergences:\n", full_table)
print("\nBest Matches for Each Dict1 Key:\n", best_matches)

# Optional: Save to CSV
# full_table.to_csv("kl_all_pairs.csv", index=False)
# best_matches.to_csv("kl_best_match.csv", index=False)
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
from scipy.optimize import curve_fit
from scipy.optimize import root_scalar


# # Create 3D scatter plot
fig = plt.figure(figsize=(9, 6))
ax = fig.add_subplot(111, projection='3d')

# ax.plot_surface(full_t, full_ee, full_steepness, cmap='viridis', alpha=0.5)
ax.scatter(filtered_aa_temperature_list, filtered_all_ee_list, filtered_z2d, c='red', s=100) 


def poly2d(x, a0, a1, a2):
    return a0 + a1 * x + a2 * x*x

def exp_fit(x,a0,b0):
    return b0*np.exp(a0*x)

def sigmoid_fit(x, a0, b0, c0):
    return a0 / (1 + np.exp(b0 * (x + c0)))

def inv_sigmoid(x, a0, b0, c0):
    return 4.6 + a0 / (1 + np.exp(b0 * (x - c0)))

def sigmoid(x, L, x0, k, b):
    return L / (1 + np.exp(-k*(x - x0))) + b

def exp_decay(x, a, b, c):
    return a * np.exp(-b * x) + c

# Fit the plane
# popt, pcov = curve_fit(poly2d, (filtered_all_ee_list), filtered_z2d)
# t_popt, t_pcov = curve_fit(poly2d, (filtered_aa_temperature_list), filtered_z2d)

yerr = np.array(filtered_z2d_lower)-np.array(filtered_z2d_lower)
yerr[yerr == 0] = 1e-6

popt, pcov = curve_fit(exp_fit, (filtered_all_ee_list), filtered_z2d, sigma=yerr , absolute_sigma=True)
t_popt, t_pcov = curve_fit(inv_sigmoid, (filtered_aa_temperature_list), filtered_z2d, sigma=yerr, absolute_sigma=True)


print(f'EE-Steepness --- a0:{popt[0]}, a1:{popt[1]}') #, a2:{popt[2]}')
print(f'Temperature-Steepness --- a0:{t_popt[0]}, a1:{t_popt[1]}, a2:{t_popt[2]}')#, a3:{t_popt[3]}')

t_new = np.array(filtered_aa_temperature_list)
ee_new = np.array(filtered_all_ee_list)
# ee_pred = poly2d((ee_new), *popt)
ee_pred = exp_fit((ee_new), *popt)

ax.plot(t_new, ee_new, ee_pred, color='red', lw=2)
# temp_pred = poly2d((t_new), *t_popt)
temp_pred = inv_sigmoid((t_new), *t_popt)

ax.plot(t_new, ee_new, temp_pred, color='black', lw=2)

# # Predict for new values
t_star = np.arange(2.,6,0.2) #np.array(filtered_aa_temperature_list)

# def find_positive_root(y, a0, a1, a2, x_min=1e-6, x_max=100, steps=10000):
#     x_vals = np.linspace(x_min, x_max, steps)
#     f_vals = poly2d(x_vals, a0, a1, a2) - y
    
#     for i in range(len(f_vals) - 1):
#         if f_vals[i] * f_vals[i+1] < 0:  # sign change
#             bracket = (x_vals[i], x_vals[i+1])
#             result = root_scalar(lambda x: poly2d(x, a0, a1, a2) - y, bracket=bracket, method='brentq')
#             if result.converged:
#                 return result.root
#     return None  # No root found in this interval

temp_star_pred = inv_sigmoid((t_star), *t_popt)
# ee_star_pred = np.array([find_positive_root(x, *popt) for x in temp_star_pred])
# print(t_new)
# print(filtered_z2d)
# print(ee_new)

# print(t_star)
# print(temp_star_pred)
# print(ee_star_pred)
# print(temp_pred.shape, ee_pred.shape)

# ax.plot(t_star, ee_star_pred, temp_star_pred, color='green', lw=2)

# ee_new

```
