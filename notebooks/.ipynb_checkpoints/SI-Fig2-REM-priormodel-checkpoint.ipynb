```python
################################################################################################

rdf_results_file_max_6 = traj_file_path +  'chiral_results/nosymmetry_rdf_halfbox_maxbound_8_nbinary.pkl'

with open(rdf_results_file_max_6, 'rb') as file:
    ucg_comRDF_max_6 = pickle.load(file)
        
```

```python
temperature = 4.6
kt_reduced = temperature
```

```python
fig = plt.figure(figsize=(10, 5))

plt.subplots_adjust(left=0.1, right=0.9, bottom=0.1, top=0.9, wspace=0.35, hspace=0.35)

grid = GridSpec(1, 2)

ax1 = plt.subplot(grid[0, 0])
ax2 = plt.subplot(grid[0, 1])
# ax3 = plt.subplot(grid[1, 0])


# Create subplots
################################################################################################
ucg_comRDF = ucg_comRDF_max_6 ## ENABLE FOR CUTOFF 6


################################################################################################
################################################################################################
################################################################################################

ax1.scatter(ucg_comRDF['t1']['bin_centers'], ucg_comRDF['t1']['rdf']*2, color='blue', s=2)
ax1.plot(ucg_comRDF['t1']['bin_centers'], ucg_comRDF['t1']['rdf']*2, color='blue', lw=2, label = 'L-L')

ax1.scatter(ucg_comRDF['t2']['bin_centers'], ucg_comRDF['t2']['rdf']*2, color='green', s=2)
ax1.plot(ucg_comRDF['t2']['bin_centers'], ucg_comRDF['t2']['rdf']*2, color='green', lw=2, label = 'D-D')

ax1.scatter(ucg_comRDF['t12']['bin_centers'], ucg_comRDF['t12']['rdf']*1, color='red', s=5)
ax1.plot(ucg_comRDF['t12']['bin_centers'], ucg_comRDF['t12']['rdf']*1, color='red', lw=2, label = 'L-D')


ax1.set_xlabel(r"$r$")
ax1.set_ylabel("g(r)")

# ax1.set_xlim(0,6)
# ax1.set_ylim(-0.1,1.6)
# ax1.set_ylim(-0.02,.8)

ax1.grid()
ax1.legend(frameon=False, fontsize=12)

ax1.xaxis.set_major_locator(MultipleLocator(1))
ax1.yaxis.set_major_locator(MultipleLocator(.5))
ax1.xaxis.set_minor_locator(AutoMinorLocator(n=5))
ax1.yaxis.set_minor_locator(AutoMinorLocator(n=5))

################################################################################################
################################################################################################
################################################################################################

#prior potentials
ull_pot = -kt_reduced*np.log(ucg_comRDF['t1']['rdf']*2) + kt_reduced*np.log(ucg_comRDF['t1']['rdf']*2)[-1]
udd_pot = -kt_reduced*np.log(ucg_comRDF['t2']['rdf']*2) + kt_reduced*np.log(ucg_comRDF['t2']['rdf']*2)[-1]
uld_pot = -kt_reduced*np.log(ucg_comRDF['t12']['rdf']*1) + kt_reduced*np.log(ucg_comRDF['t12']['rdf']*1)[-1]


################################################################################################
################################################################################################
################################################################################################

target_folder = '/project2/andrewferguson/sivadasetty/doe/ucg/notebooks/chiral/manuscript/simulations/rem/high-t-repeat-nneigh12-NR-Hess-7-EXTENDED/master/init_potentials_4.6/'

saved_pot_ll = pd.read_csv(target_folder + '/ibi_guess/prior_model/Pair_L-L.table', skiprows=4, 
                                   sep='\s+', names=['r', 'u', 'f'])
saved_pot_dd = pd.read_csv(target_folder + '/ibi_guess/prior_model/Pair_D-D.table', skiprows=4, 
                                   sep='\s+', names=['r', 'u', 'f'])
saved_pot_ld = pd.read_csv(target_folder + '/ibi_guess/prior_model/Pair_L-D.table', skiprows=4, 
                                   sep='\s+', names=['r', 'u', 'f'])

################################################################################################
################################################################################################
################################################################################################


target_folder = '/project2/andrewferguson/sivadasetty/doe/ucg/notebooks/chiral/manuscript/simulations/rem/high-t-repeat-nneigh12-NR-Hess-7-EXTENDED/master/init_potentials_4.6/'

saved_pot_ll = pd.read_csv(target_folder + '/iter_0/Pair_L-L.table', skiprows=4, 
                                   sep='\s+', names=['r', 'u', 'f'])
saved_pot_dd = pd.read_csv(target_folder + '/iter_0/Pair_D-D.table', skiprows=4, 
                                   sep='\s+', names=['r', 'u', 'f'])
saved_pot_ld = pd.read_csv(target_folder + '/iter_0/Pair_L-D.table', skiprows=4, 
                                   sep='\s+', names=['r', 'u', 'f'])


################################################################################################
################################################################################################
################################################################################################


ax2.plot(ucg_comRDF['t1']['bin_centers'], ull_pot, color='blue', lw=2, label = 'L-L', alpha=0.5)
# ax2.plot(ucg_comRDF['t2']['bin_centers'], udd_pot, color='green', lw=2, label = 'D-D', alpha=0.5)
ax2.plot(ucg_comRDF['t12']['bin_centers'], uld_pot, color='red', lw=2, label = 'L-D', alpha=0.5)

################################################################################################
################################################################################################
################################################################################################

ax2.scatter(saved_pot_ll.r, saved_pot_ll.u, color='blue', s=20)
# ax2.scatter(saved_pot_dd.r, saved_pot_dd.u, color='green', s=10)
ax2.scatter(saved_pot_ld.r, saved_pot_ld.u, color='red', s=20)

################################################################################################
################################################################################################
################################################################################################

ax2.plot(saved_pot_ll.r, saved_pot_ll.u, color='blue', lw=2)
# ax2.plot(saved_pot_dd.r, saved_pot_dd.u, color='green', lw=2)
ax2.plot(saved_pot_ld.r, saved_pot_ld.u, color='red', lw=2)

ax2.grid()
# ax2.set_ylim(-5,60)

ax2.set_xlabel(r"$r$")
ax2.set_ylabel("U($r$)")

# ax2.xaxis.set_major_locator(MultipleLocator(1))
# ax2.yaxis.set_major_locator(MultipleLocator(10))
# ax2.xaxis.set_minor_locator(AutoMinorLocator(n=5))
# # ax2.yaxis.set_minor_locator(AutoMinorLocator(n=5))
 
################################################################################################
################################################################################################
################################################################################################

# Create inset of mixed specifications in the second subplot;
# width is 30% of parent Axes' bounding box and
# height is 1 inch at the upper left corner.
axins3 = inset_axes(ax2, width="60%", height=1.5, loc="upper right")

axins3.plot(ucg_comRDF['t1']['bin_centers'], ull_pot, color='blue', lw=2, label = 'HO', alpha=0.5)
# axins3.plot(ucg_comRDF['t2']['bin_centers'], udd_pot, color='green', lw=2, label = 'D-D', alpha=0.5)
axins3.plot(ucg_comRDF['t12']['bin_centers'], uld_pot, color='red', lw=2, label = 'HE', alpha=0.5)

axins3.scatter(saved_pot_ll.r, saved_pot_ll.u, color='blue', s=20)
# axins3.scatter(saved_pot_dd.r, saved_pot_dd.u, color='green', s=10)
axins3.scatter(saved_pot_ld.r, saved_pot_ld.u, color='red', s=20)

axins3.plot(saved_pot_ll.r, saved_pot_ll.u, color='blue', lw=2)
# axins3.plot(saved_pot_dd.r, saved_pot_dd.u, color='green', lw=2)
axins3.plot(saved_pot_ld.r, saved_pot_ld.u, color='red', lw=2)

axins3.set_ylim(-2.5,5)
axins3.grid()

axins3.set_xlabel(r"$r$")
axins3.set_ylabel("U($r$)")

axins3.xaxis.set_major_locator(MultipleLocator(1))
axins3.yaxis.set_major_locator(MultipleLocator(2))
axins3.xaxis.set_minor_locator(AutoMinorLocator(n=5))
# ax2.yaxis.set_minor_locator(AutoMinorLocator(n=5))

axins3.legend(frameon=False)

# #####################################################################################
for i, ax in enumerate([ax1, ax2]):#, ax3, ax4]):
    ax.text(-0.2, 1.05, '(' + string.ascii_lowercase[i] + ')', transform=ax.transAxes, size=18, weight='normal', va='top', ha='center')

```
