## Compare chiral counts

```python
shift_temperature = np.linspace(0, 9, len(temperatures))


fig = plt.figure(figsize=(10, 3))
plt.subplots_adjust(left=0.1, right=0.9, bottom=0.1, top=0.9, wspace=0.35, hspace=0.2)
grid = GridSpec(1, 2)


ax1 = plt.subplot(grid[0, 0])
aa_chiral_hist_temperature = {}
aa_ee = {}
for i, temperature in enumerate(temperatures):
    ax1.plot(np.arange(len(chiralcount_max_6[temperature]['l'])), chiralcount_max_6[temperature]['l'].values(), color='blue')
    ax1.plot(np.arange(len(chiralcount_max_6[temperature]['d'])), chiralcount_max_6[temperature]['d'].values(), color='red')

    
    l_hist, l_bin_edges = np.histogram(np.array(list(chiralcount_max_6[temperature]['l'].values())), bins=50, density=True)
    d_hist, d_bin_edges = np.histogram(np.array(list(chiralcount_max_6[temperature]['d'].values())), bins=50, density=True)
    
    l_bin_centers = 0.5 * (l_bin_edges[:-1] + l_bin_edges[1:])
    d_bin_centers = 0.5 * (d_bin_edges[:-1] + d_bin_edges[1:])
    
    aa_chiral_hist_temperature[temperature] = {'l_hist': l_hist, 'l_bin_edges': l_bin_edges, 'l_bin_centers': l_bin_centers,
                                               'd_hist': d_hist, 'd_bin_edges': d_bin_edges, 'd_bin_centers': d_bin_centers}
    
    numerator = np.array(list(chiralcount_max_6[temperature]['l'].values()))-np.array(list(chiralcount_max_6[temperature]['d'].values()))
    denominator = np.array(list(chiralcount_max_6[temperature]['l'].values()))+np.array(list(chiralcount_max_6[temperature]['d'].values()))
    aa_ee[temperature] = np.array(numerator)/np.array(denominator)
    

ax1.set_xlabel(r"$Frame$")
ax1.set_ylabel("Count")
    
ax1.xaxis.set_major_locator(MultipleLocator(4000))
ax1.yaxis.set_major_locator(MultipleLocator(200))
ax1.xaxis.set_minor_locator(AutoMinorLocator(n=5))
ax1.yaxis.set_minor_locator(AutoMinorLocator(n=5))
ax1.grid()

ax2 = plt.subplot(grid[0, 1])
bi_ucg_chiral_hist_temperature = {}
bi_ee = {}
for i, temperature in enumerate(temperatures):
    ax2.plot(np.arange(len(bi_ucg_chiralcount_max_6[temperature]['l'])), bi_ucg_chiralcount_max_6[temperature]['l'].values(), color='blue')
    ax2.plot(np.arange(len(bi_ucg_chiralcount_max_6[temperature]['d'])), bi_ucg_chiralcount_max_6[temperature]['d'].values(), color='red')

    l_hist, l_bin_edges = np.histogram(np.array(list(bi_ucg_chiralcount_max_6[temperature]['l'].values())), bins=50, density=True)
    d_hist, d_bin_edges = np.histogram(np.array(list(bi_ucg_chiralcount_max_6[temperature]['d'].values())), bins=50, density=True)
    
    l_bin_centers = 0.5 * (l_bin_edges[:-1] + l_bin_edges[1:])
    d_bin_centers = 0.5 * (d_bin_edges[:-1] + d_bin_edges[1:])
    
    bi_ucg_chiral_hist_temperature[temperature] = {'l_hist': l_hist, 'l_bin_edges': l_bin_edges, 'l_bin_centers': l_bin_centers,
                                                   'd_hist': d_hist, 'd_bin_edges': d_bin_edges, 'd_bin_centers': d_bin_centers}
    
    numerator = (np.array(list(bi_ucg_chiralcount_max_6[temperature]['l'].values()))-np.array(list(bi_ucg_chiralcount_max_6[temperature]['d'].values())))
    denominator = (np.array(list(bi_ucg_chiralcount_max_6[temperature]['l'].values()))+np.array(list(bi_ucg_chiralcount_max_6[temperature]['d'].values())))
    bi_ee[temperature] = np.array(numerator)/np.array(denominator)

ax2.set_xlabel(r"$Frame$")
ax2.set_ylabel("Count")
        
ax2.xaxis.set_major_locator(MultipleLocator(10000))
ax2.yaxis.set_major_locator(MultipleLocator(200))
ax2.xaxis.set_minor_locator(AutoMinorLocator(n=5))
ax2.yaxis.set_minor_locator(AutoMinorLocator(n=5))
ax2.grid()
```

## EE plot

```python
fig = plt.figure(figsize=(15, 3))
    
font = {'weight' : 'normal',
        'size'   : 20}
plt.rc('font', **font)

plt.subplots_adjust(left=0.1, right=0.9, bottom=0.1, top=0.9, wspace=0.35, hspace=0.2)

grid = GridSpec(1, 3)

## aa subplots
ax1 = plt.subplot(grid[0, 0])
aa_ee_symmetry_hist_temperature = {}
for i, temperature in enumerate(temperatures):
    ax1.plot(np.arange(len(aa_ee[temperature])), aa_ee[temperature], c='blue')

    aa_ee_symmetry = np.concatenate((aa_ee[temperature], -aa_ee[temperature]))
    aa_ee_symmetry_hist, aa_ee_symmetry_bin_edges = np.histogram(aa_ee_symmetry, bins=30, range=(-1,1), density=True)
    aa_ee_symmetry_bin_centers = 0.5 * (aa_ee_symmetry_bin_edges[:-1] + aa_ee_symmetry_bin_edges[1:])
    aa_ee_symmetry_hist_temperature[temperature] = {'ee_symmetry_hist': aa_ee_symmetry_hist, 'ee_symmetry_bin_edges': aa_ee_symmetry_bin_edges, 'ee_symmetry_bin_centers': aa_ee_symmetry_bin_centers}
    
    
ax1.grid()
ax1.set_xlabel("Frame")
ax1.set_ylabel("EE")
ax1.set_ylim(-1,1)
ax1.grid()

ax3 = plt.subplot(grid[0, 1])
bi_ee_symmetry_hist_temperature = {}
for i, temperature in enumerate(temperatures):
    ax3.plot(np.arange(len(bi_ee[temperature])), bi_ee[temperature], c='red')

    bi_ee_symmetry = np.concatenate((bi_ee[temperature], -bi_ee[temperature]))
    bi_ee_symmetry_hist, bi_ee_symmetry_bin_edges = np.histogram(bi_ee_symmetry, bins=30, range=(-1,1), density=True)
    bi_ee_symmetry_bin_centers = 0.5 * (bi_ee_symmetry_bin_edges[:-1] + bi_ee_symmetry_bin_edges[1:])
    bi_ee_symmetry_hist_temperature[temperature] = {'ee_symmetry_hist': bi_ee_symmetry_hist, 'manual_ee_symmetry_bin_edges': bi_ee_symmetry_bin_edges, 'ee_symmetry_bin_centers': bi_ee_symmetry_bin_centers}

ax3.grid()
ax3.set_xlabel("Frame")
ax3.set_ylabel("EE")
ax3.set_ylim(-1,1)
ax3.grid()

```
