## Compare RDFs

```python

fig = plt.figure(figsize=(16, 4))
plt.subplots_adjust(left=0.1, right=0.9, bottom=0.1, top=0.9, wspace=0.35, hspace=0.2)
grid = GridSpec(1, 3)

ax1 = plt.subplot(grid[0, 0])

ax1.plot(aa_RDF_max_6['t1']['bin_centers'], aa_RDF_max_6['t1']['rdf']*2 + 0, color='blue')
ax1.plot(new_aa_RDF_max_6['t1']['bin_centers'], new_aa_RDF_max_6['t1']['rdf']*2 + 0, color='red')

ax1.set_xlabel(r"$r$")
ax1.set_ylabel("g($r$)")
        
ax1.xaxis.set_major_locator(MultipleLocator(1))
ax1.yaxis.set_major_locator(MultipleLocator(.5))
ax1.xaxis.set_minor_locator(AutoMinorLocator(n=5))
ax1.yaxis.set_minor_locator(AutoMinorLocator(n=5))
ax1.set_ylim(0.,1.5)
ax1.grid()

ax1 = plt.subplot(grid[0, 1])

ax1.plot(aa_RDF_max_6['t2']['bin_centers'], aa_RDF_max_6['t2']['rdf']*2 + 0, color='blue')
ax1.plot(new_aa_RDF_max_6['t2']['bin_centers'], new_aa_RDF_max_6['t2']['rdf']*2 + 0, color='red')


ax1.set_xlabel(r"$r$")
ax1.set_ylabel("g($r$)")
        
ax1.xaxis.set_major_locator(MultipleLocator(1))
ax1.yaxis.set_major_locator(MultipleLocator(.5))
ax1.xaxis.set_minor_locator(AutoMinorLocator(n=5))
ax1.yaxis.set_minor_locator(AutoMinorLocator(n=5))
ax1.set_ylim(0.,1.5)
ax1.grid()

ax2 = plt.subplot(grid[0, 2])

ax2.plot(aa_RDF_max_6['t12']['bin_centers'], aa_RDF_max_6['t12']['rdf']*1 + 0, color='blue')
ax2.plot(new_aa_RDF_max_6['t12']['bin_centers'], new_aa_RDF_max_6['t12']['rdf']*1 + 0, color='red')

ax2.set_xlabel(r"$r$")
ax2.set_ylabel("g($r$)")
    
ax2.xaxis.set_major_locator(MultipleLocator(1))
ax1.yaxis.set_major_locator(MultipleLocator(.5))
ax2.xaxis.set_minor_locator(AutoMinorLocator(n=5))
ax2.yaxis.set_minor_locator(AutoMinorLocator(n=5))
ax2.set_ylim(0.,1.5)
ax2.grid()

```
