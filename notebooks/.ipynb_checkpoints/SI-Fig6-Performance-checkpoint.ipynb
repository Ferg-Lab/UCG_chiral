## Performance

```python
aa_speed = {}
aa_speed_mean = []
aa_speed_std = []

for size in [1000, 2000, 4000, 8000]:

    aa_speed[size] = pd.read_csv('/project2/andrewferguson/sivadasetty/doe/ucg/chiral_tetramer/aa-performance/'+str(size) + '/speed.txt', header=None, names=['speed'])
    
    aa_speed_mean.append(aa_speed[size]['speed'].values.mean())
    aa_speed_std.append(aa_speed[size]['speed'].values.std())


```

```python
ucg_speed = {}
ucg_speed_mean = []
ucg_speed_std = []

for size in [1000, 2000, 4000, 8000]:

    ## NOTE: These are run at old potentials corresponding to high-t-repeat-nneigh12-NR-Hess-7-not-parallel in prelim runs folder
    ucg_speed[size] = pd.read_csv('/project2/andrewferguson/sivadasetty/doe/ucg/notebooks/chiral/manuscript/simulations/rem/high-t-repeat-nneigh12-NR-Hess-7-EXTENDED/performance-tests/size_'+str(size) + '/speed.txt', header=None, names=['speed'])

    ucg_speed_mean.append(ucg_speed[size]['speed'].values.mean())
    ucg_speed_std.append(ucg_speed[size]['speed'].values.std())
    
    
```

## ucg_speed

```python


fig = plt.figure(figsize=(8, 6))
plt.subplots_adjust(left=0.1, right=0.9, bottom=0.1, top=0.9, wspace=0.2, hspace=0.35)
grid = GridSpec(1, 1)

colors=['darkblue', 'darkorange','darkgreen','darkred']

ax1 = plt.subplot(grid[0,0])

ax1.scatter(list(aa_speed.keys()), aa_speed_mean, c='blue', alpha=1., lw=3)
ax1.errorbar(list(aa_speed.keys()), aa_speed_mean, yerr=aa_speed_std, c='blue', alpha=1., lw=3, label='AA')

ax1.scatter(list(ucg_speed.keys()), ucg_speed_mean, c='red', alpha=1., lw=3)
ax1.errorbar(list(ucg_speed.keys()), ucg_speed_mean, yerr=ucg_speed_std, c='red', alpha=1., lw=3, label='UCG')


ax1.set_ylabel("Time (s)")
ax1.set_xlabel("N")
ax1.grid()
# ax2.xaxis.set_major_locator(MultipleLocator(.5))
ax1.yaxis.set_major_locator(MultipleLocator(25))
# ax2.xaxis.set_minor_locator(AutoMinorLocator(n=5))
ax1.yaxis.set_minor_locator(AutoMinorLocator(n=5))
# ax1.set_ylim(-1.,1.)# ax1.text(4100, 0.65, r'N: ' + str(size), fontsize=12, color=colors[k])# if k == 0:
# ax1.set_title('T:' + str(temperature))

ax1.legend(frameon=False, loc='upper left')

```
