## Observables

```python
def kl_divergence(p, q, eps=1e-12):
    
    p = np.asarray(p, dtype=np.float64)
    q = np.asarray(q, dtype=np.float64)
    p /= np.sum(p)
    q /= np.sum(q)
    #p = np.clip(p, eps, 1)
    #q = np.clip(q, eps, 1)
    return np.sum(p * np.log(p / q))

def rdf_fit(p, q):

    # p = p[skip:]
    # q = q[skip:]
    
    diff_t1 =  np.sum(np.abs(q-p))
    diff_t1 /= np.sum(np.abs(q)+np.abs(p))

    return 1-diff_t1
```

## EE calculations

```python

aa_chiral_hist_temperature = {}
aa_ee = {}

l_hist, l_bin_edges = np.histogram(np.array(list(chiralcount_max_6[temperature]['l'].values())[::1]), bins=50, density=True)
d_hist, d_bin_edges = np.histogram(np.array(list(chiralcount_max_6[temperature]['d'].values())[::1]), bins=50, density=True)
    
l_bin_centers = 0.5 * (l_bin_edges[:-1] + l_bin_edges[1:])
d_bin_centers = 0.5 * (d_bin_edges[:-1] + d_bin_edges[1:])
    
aa_chiral_hist_temperature[temperature] = {'l_hist': l_hist, 'l_bin_edges': l_bin_edges, 'l_bin_centers': l_bin_centers,
                                               'd_hist': d_hist, 'd_bin_edges': d_bin_edges, 'd_bin_centers': d_bin_centers}

skip=1
numerator = np.array(list(chiralcount_max_6[temperature]['l'].values()))[::skip]-np.array(list(chiralcount_max_6[temperature]['d'].values()))[::skip]
denominator = np.array(list(chiralcount_max_6[temperature]['l'].values()))[::skip]+np.array(list(chiralcount_max_6[temperature]['d'].values()))[::skip]
aa_ee[temperature] = np.array(numerator)/np.array(denominator)
    


bi_ucg_chiral_hist_temperature = {}
bi_ee = {}

bi_ucg_chiral_hist_temperature_iter = {}
bi_ee_iter = {}

for k,iter in enumerate(list(bi_ucg_chiralcount_max_6[temperature].keys())):
    
    l_hist, l_bin_edges = np.histogram(np.array(list(bi_ucg_chiralcount_max_6[temperature][iter]['l'].values())), bins=50, density=True)
    d_hist, d_bin_edges = np.histogram(np.array(list(bi_ucg_chiralcount_max_6[temperature][iter]['d'].values())), bins=50, density=True)
        
    l_bin_centers = 0.5 * (l_bin_edges[:-1] + l_bin_edges[1:])
    d_bin_centers = 0.5 * (d_bin_edges[:-1] + d_bin_edges[1:])
        
    bi_ucg_chiral_hist_temperature_iter[iter] = {'l_hist': l_hist, 'l_bin_edges': l_bin_edges, 'l_bin_centers': l_bin_centers,
                                                 'd_hist': d_hist, 'd_bin_edges': d_bin_edges, 'd_bin_centers': d_bin_centers}

    skip=1
    #print(np.array(list(bi_ucg_chiralcount_max_6[temperature][iter]['l'].values()))[::skip].shape)
    numerator = (np.array(list(bi_ucg_chiralcount_max_6[temperature][iter]['l'].values()))[::skip]-np.array(list(bi_ucg_chiralcount_max_6[temperature][iter]['d'].values()))[::skip])
    denominator = (np.array(list(bi_ucg_chiralcount_max_6[temperature][iter]['l'].values()))[::skip]+np.array(list(bi_ucg_chiralcount_max_6[temperature][iter]['d'].values()))[::skip])
    bi_ee_iter[iter] = np.array(numerator)/np.array(denominator)

bi_ucg_chiral_hist_temperature[temperature] = bi_ucg_chiral_hist_temperature_iter
bi_ee[temperature] = bi_ee_iter

```
