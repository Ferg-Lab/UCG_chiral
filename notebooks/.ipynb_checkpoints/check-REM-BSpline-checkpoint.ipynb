```python
import numpy as np
from typing import Iterable, Dict, Tuple

import numpy as np
from functools import lru_cache
from typing import Dict

# ---------- Pair-index cache (avoids storing 'pairs') ----------
@lru_cache(maxsize=None)
def _pair_index_upper(N: int):
    """
    Returns (i_idx, j_idx) of length M = N*(N-1)//2 in pdist/upper-tri order.
    Cached per N for large trajectories.
    """
    i_idx, j_idx = np.triu_indices(N, k=1)
    return i_idx.astype(np.int32), j_idx.astype(np.int32)


# ---------- Make w_like / w_unlike from per-particle p ----------
def pair_state_weights_from_p(p: np.ndarray, N: int):
    """
    Inputs:
      p : (N,) probabilities for s=1 per particle
      N : number of particles (len(p))
    Returns:
      w_like, w_unlike : (M,) per-pair weights in upper-tri 'pdist' order
    """
    i_idx, j_idx = _pair_index_upper(N)
    pi = p[i_idx]
    pj = p[j_idx]
    w_unlike = pi + pj - 2.0 * pi * pj
    w_like = 1.0 - w_unlike
    return w_like, w_unlike


# ---------- Per-frame feature sums using only (R, p) ----------
def frame_feature_sums_from_p(
    R: np.ndarray,
    p: np.ndarray,
    model_like: "SplineModel",
    model_unlike: "SplineModel",
) -> Dict[str, np.ndarray]:
    """
    R : (M,) pair distances in condensed upper-tri order for this frame
    p : (N,) per-particle state-1 probabilities for this frame
    Returns dict with 'f_like', 'f_unlike', 'E_like', 'E_unlike'
    """
    N = p.shape[0]
    # Build per-pair weights from per-particle probabilities
    w_like, w_unlike = pair_state_weights_from_p(p, N)

    # Design matrices at distances
    Phi_like   = model_like.jacobian_wrt_coeffs(R)       # (M, nb_like)
    Phi_unlike = model_unlike.jacobian_wrt_coeffs(R)     # (M, nb_unlike)

    # Weighted feature sums (sufficient stats for REM)
    f_like   = (w_like[:, None]   * Phi_like).sum(axis=0)
    f_unlike = (w_unlike[:, None] * Phi_unlike).sum(axis=0)

    #print(f'debug {f_like} {model_like.c} {f_unlike} {model_unlike.c}')

    E_like   = float(f_like @ model_like.c)    if model_like.c    is not None else 0.0
    E_unlike = float(f_unlike @ model_unlike.c) if model_unlike.c is not None else 0.0

    return {"f_like": f_like, "f_unlike": f_unlike, "E_like": E_like, "E_unlike": E_unlike}


# ---------- Ensemble means over frames: each has only (R, p) ----------
def ensemble_feature_means_from_p(
    frames,                        # iterable of {"R": (M,), "p": (N,)}
    model_like: "SplineModel",
    model_unlike: "SplineModel",
) -> Dict[str, np.ndarray]:
    """
    Streams large trajectories without storing per-pair arrays.
    """
    nb_like   = model_like.nbasis
    nb_unlike = model_unlike.nbasis
    acc_f_like   = np.zeros(nb_like)
    acc_f_unlike = np.zeros(nb_unlike)
    acc_E_like = 0.0
    acc_E_unlike = 0.0
    n = 0

    use_upper_triangle_only = True


    for f, fr in enumerate(frames[::100]):
        
        if use_upper_triangle_only:
            mask = np.triu(np.ones_like(fr['R'], dtype=bool), k=1)
        else:
            mask = ~np.eye(N, dtype=bool)
        rij = fr['R'][mask]
        out = frame_feature_sums_from_p(rij, fr["p"], model_like, model_unlike)
        acc_f_like   += out["f_like"]
        acc_f_unlike += out["f_unlike"]
        acc_E_like   += out["E_like"]
        acc_E_unlike += out["E_unlike"]

        print(f'{f} {out["E_like"]} {out["E_unlike"]}')
        n += 1

    if n == 0:
        raise ValueError("No frames provided to ensemble_feature_means_from_p().")

    return {
        "mean_f_like":   acc_f_like / n,
        "mean_f_unlike": acc_f_unlike / n,
        "mean_E_like":   acc_E_like / n,
        "mean_E_unlike": acc_E_unlike / n,
        "n_frames":      n,
    }


```

```python
# ---------- REM objective (linear in coefficients) ----------
def rem_objective_and_grad(
    aa_stats: Dict[str, np.ndarray],
    cg_stats: Dict[str, np.ndarray],
    c_like: np.ndarray,
    c_unlike: np.ndarray,
    reg_l2: float = 0.0,
) -> Tuple[float, np.ndarray]:
    """
    Relative Entropy Minimization (up to an additive constant):
      L(c) = <U_mix(c)>_CG - <U_mix(c)>_AA  + (reg_l2/2) * ||c||^2
    where U_mix(c) = c_like·f_like + c_unlike·f_unlike.
    Inputs:
      - aa_stats, cg_stats: outputs of ensemble_feature_means() for AA and CG ensembles
      - c_like, c_unlike: current coefficient vectors
      - reg_l2: optional L2 regularization (same weight for both blocks)
    Returns:
      (loss, grad) where grad concatenates [grad_like, grad_unlike]
    """
    # feature means (sufficient statistics)
    d_f_like   = cg_stats["mean_f_like"]   - aa_stats["mean_f_like"]
    d_f_unlike = cg_stats["mean_f_unlike"] - aa_stats["mean_f_unlike"]

    # loss
    loss = float(d_f_like @ c_like + d_f_unlike @ c_unlike)
    if reg_l2 > 0.0:
        loss += 0.5 * reg_l2 * (np.dot(c_like, c_like) + np.dot(c_unlike, c_unlike))

    # gradient wrt coefficients
    grad_like = d_f_like + (reg_l2 * c_like if reg_l2 > 0.0 else 0.0)
    grad_unlike = d_f_unlike + (reg_l2 * c_unlike if reg_l2 > 0.0 else 0.0)

    grad = np.concatenate([grad_like, grad_unlike])
    return loss, grad


# ---- one gradient step using your rem_objective_and_grad ----
def rem_step(c_like, c_unlike, aa_stats, cg_stats, reg_l2=0.0, lr=1e-2):
    loss, g = rem_objective_and_grad(aa_stats, cg_stats, c_like, c_unlike, reg_l2=reg_l2)
    nb_like = c_like.size
    g_like, g_unlike = g[:nb_like], g[nb_like:]
    c_like_new   = c_like   - lr * g_like
    c_unlike_new = c_unlike - lr * g_unlike
    return c_like_new, c_unlike_new, loss


# ---- tiny training loop (you can stop when loss plateaus) ----
def fit_rem_gd(c_like, c_unlike, aa_stats, cg_stats, reg_l2=0.0, lr=1e-2, iters=200):
    for _ in range(iters):
        c_like, c_unlike, loss = rem_step(c_like, c_unlike, aa_stats, cg_stats, reg_l2, lr)
    return c_like, c_unlike, loss


# ---- commit updated coefficients to your spline models and evaluate ----
def update_models_and_eval(model_like, model_unlike, c_like, c_unlike, x_eval):
    # write coefficients into the models (assumes SplineModel.unpack_params exists)
    model_like.unpack_params(c_like)
    model_unlike.unpack_params(c_unlike)
    # evaluate predicted pair energy contributions (design-matrix linearity)
    y_like   = model_like.eval(x_eval)     # basis @ c_like
    y_unlike = model_unlike.eval(x_eval)   # basis @ c_unlike
    return y_like, y_unlike

```

```python
# initial guesses
c_like0   = model_like.c.copy()
c_unlike0 = model_unlike.c.copy()

# build AA/CG feature means once (using your from-p helpers)
aa_stats = ensemble_feature_means_from_p(aa_frames, model_like, model_unlike)
cg_stats = ensemble_feature_means_from_p(cg_frames, model_like, model_unlike)

# optimize
c_like_opt, c_unlike_opt, final_loss = fit_rem_gd(
    c_like0, c_unlike0, aa_stats, cg_stats, reg_l2=1e-3, lr=5e-3, iters=300
)

# commit and evaluate on any grid of distances
x_eval = np.linspace(0.1, 7.0, 400)
y_like, y_unlike = update_models_and_eval(model_like, model_unlike, c_like_opt, c_unlike_opt, x_eval)
```
