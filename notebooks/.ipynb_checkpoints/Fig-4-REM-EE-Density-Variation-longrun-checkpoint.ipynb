```python
def kl_divergence(p, q, eps=1e-12):
    p = np.asarray(p, dtype=np.float64)
    q = np.asarray(q, dtype=np.float64)
    p /= np.sum(p)
    q /= np.sum(q)
    p = np.clip(p, eps, 1)
    q = np.clip(q, eps, 1)
    return np.sum(p * np.log(p / q))

def rdf_fit(p, q, skip):

    p = p[skip:]
    q = q[skip:]
    
    diff_t1 =  np.sum(np.abs(q-p))
    diff_t1 /= np.sum(np.abs(q)+np.abs(p))

    return 1-diff_t1
```

```python

aa_chiral_hist_density = {}
aa_ee_density = {}

for density in [0.07, 0.09, 0.11]:
    aa_chiral_hist_temperature = {}
    aa_ee = {}
    for i, temperature in enumerate(all_temperatures):
        l_hist, l_bin_edges = np.histogram(np.array(list(chiralcount_max_6_density[density][temperature]['l'].values())), bins=50, density=True)
        d_hist, d_bin_edges = np.histogram(np.array(list(chiralcount_max_6_density[density][temperature]['d'].values())), bins=50, density=True)
        
        l_bin_centers = 0.5 * (l_bin_edges[:-1] + l_bin_edges[1:])
        d_bin_centers = 0.5 * (d_bin_edges[:-1] + d_bin_edges[1:])
        
        aa_chiral_hist_temperature[temperature] = {'l_hist': l_hist, 'l_bin_edges': l_bin_edges, 'l_bin_centers': l_bin_centers,
                                                   'd_hist': d_hist, 'd_bin_edges': d_bin_edges, 'd_bin_centers': d_bin_centers}
    
        skip=1
        numerator = np.array(list(chiralcount_max_6_density[density][temperature]['l'].values()))[::skip]-np.array(list(chiralcount_max_6_density[density][temperature]['d'].values()))[::skip]
        denominator = np.array(list(chiralcount_max_6_density[density][temperature]['l'].values()))[::skip]+np.array(list(chiralcount_max_6_density[density][temperature]['d'].values()))[::skip]
        aa_ee[temperature] = np.array(numerator)/np.array(denominator)

    aa_chiral_hist_density[density] = aa_chiral_hist_temperature
    aa_ee_density[density] = aa_ee
    

bi_ucg_chiral_hist_density = {}
bi_ee_density = {}
for density in [0.07, 0.09, 0.11]:

    bi_ucg_chiral_hist_temperature = {}
    bi_ee_temperature = {}

    for temperature in all_temperatures:
        bi_ucg_chiral_hist_temperature_nsteps = {}
        bi_ee_temperature_nsteps = {}
        
        for nsteps in all_nsteps:
            bi_ucg_chiral_hist_temperature_nsteps_chiral_lag = {}
            bi_ee_temperature_nsteps_chiral_lag = {}
            
            for chiral_lag in all_chiral_lag:
                bi_ucg_chiral_hist_temperature_nsteps_chiral_lag_iter = {}
                bi_ee_temperature_nsteps_chiral_lag_iter = {}
                
                for i in bi_ucg_chiral_count_density[density][temperature][nsteps][chiral_lag].keys():
                    
                    l_hist, l_bin_edges = np.histogram(np.array(list(bi_ucg_chiral_count_density[density][temperature][nsteps][chiral_lag][i]['l'].values())), bins=50, density=True)
                    d_hist, d_bin_edges = np.histogram(np.array(list(bi_ucg_chiral_count_density[density][temperature][nsteps][chiral_lag][i]['d'].values())), bins=50, density=True)
            
                    l_bin_centers = 0.5 * (l_bin_edges[:-1] + l_bin_edges[1:])
                    d_bin_centers = 0.5 * (d_bin_edges[:-1] + d_bin_edges[1:])
                    
                    bi_ucg_chiral_hist_temperature_nsteps_chiral_lag_iter[i] = {'l_hist': l_hist, 'l_bin_edges': l_bin_edges, 'l_bin_centers': l_bin_centers,
                                                                                'd_hist': d_hist, 'd_bin_edges': d_bin_edges, 'd_bin_centers': d_bin_centers}
    
                    skip=1
                    numerator = (np.array(list(bi_ucg_chiral_count_density[density][temperature][nsteps][chiral_lag][i]['l'].values()))[::skip]-np.array(list(bi_ucg_chiral_count_density[density][temperature][nsteps][chiral_lag][i]['d'].values()))[::skip])
                    denominator = (np.array(list(bi_ucg_chiral_count_density[density][temperature][nsteps][chiral_lag][i]['l'].values()))[::skip]+np.array(list(bi_ucg_chiral_count_density[density][temperature][nsteps][chiral_lag][i]['d'].values()))[::skip])
                    bi_ee_temperature_nsteps_chiral_lag_iter[i] = np.array(numerator)/np.array(denominator)
    
                bi_ucg_chiral_hist_temperature_nsteps_chiral_lag[chiral_lag] = bi_ucg_chiral_hist_temperature_nsteps_chiral_lag_iter
                bi_ee_temperature_nsteps_chiral_lag[chiral_lag] = bi_ee_temperature_nsteps_chiral_lag_iter
    
            bi_ucg_chiral_hist_temperature_nsteps[nsteps] = bi_ucg_chiral_hist_temperature_nsteps_chiral_lag
            bi_ee_temperature_nsteps[nsteps] = bi_ee_temperature_nsteps_chiral_lag
    
        bi_ucg_chiral_hist_temperature[temperature] = bi_ucg_chiral_hist_temperature_nsteps
        bi_ee_temperature[temperature] = bi_ee_temperature_nsteps 
        
    bi_ucg_chiral_hist_density[density] = bi_ucg_chiral_hist_temperature
    bi_ee_density[density] = bi_ee_temperature

```

```python

aa_ee_symmetry_hist_density = {}

for density in [0.07, 0.09, 0.11]:
    
    aa_ee_symmetry_hist_temperature = {}
    for i, temperature in enumerate(all_temperatures):
    
        aa_ee_symmetry = np.concatenate((aa_ee_density[density][temperature], -aa_ee_density[density][temperature]))
        aa_ee_symmetry_hist, aa_ee_symmetry_bin_edges = np.histogram(aa_ee_symmetry, bins=30, range=(-1,1), density=True)
        aa_ee_symmetry_bin_centers = 0.5 * (aa_ee_symmetry_bin_edges[:-1] + aa_ee_symmetry_bin_edges[1:])
        aa_ee_symmetry_hist_temperature[temperature] = {'ee_symmetry_hist': aa_ee_symmetry_hist/np.sum(aa_ee_symmetry_hist), 'ee_symmetry_bin_edges': aa_ee_symmetry_bin_edges, 'ee_symmetry_bin_centers': aa_ee_symmetry_bin_centers}
        
    aa_ee_symmetry_hist_density[density] = aa_ee_symmetry_hist_temperature


bi_ee_symmetry_hist_density = {}

for density in [0.07, 0.09, 0.11]:
    
    bi_ee_symmetry_hist_temperature = {}
    
    for temperature in all_temperatures:
        bi_ee_symmetry_hist_temperature_nsteps = {}
        
        for nsteps in all_nsteps:
            bi_ee_symmetry_hist_temperature_nsteps_chiral_lag = {}
            
            for chiral_lag in all_chiral_lag:
                bi_ee_symmetry_hist_temperature_nsteps_chiral_lag_iter = {}
                
                for i in bi_ee_density[density][temperature][nsteps][chiral_lag].keys():
    
                    bi_ee_symmetry = np.concatenate((bi_ee_density[density][temperature][nsteps][chiral_lag][i], - bi_ee_density[density][temperature][nsteps][chiral_lag][i]))

                    #print(i, bi_ee_symmetry)
                    
                    bi_ee_symmetry_hist, bi_ee_symmetry_bin_edges = np.histogram(bi_ee_symmetry, bins=30, range=(-1,1), density=False)
                    bi_ee_symmetry_bin_centers = 0.5 * (bi_ee_symmetry_bin_edges[:-1] + bi_ee_symmetry_bin_edges[1:])
                    bi_ee_symmetry_hist_temperature_nsteps_chiral_lag_iter[i] = {'ee_symmetry_hist': bi_ee_symmetry_hist/np.sum(bi_ee_symmetry_hist), 'manual_ee_symmetry_bin_edges': bi_ee_symmetry_bin_edges, 'ee_symmetry_bin_centers': bi_ee_symmetry_bin_centers}
                
                bi_ee_symmetry_hist_temperature_nsteps_chiral_lag[chiral_lag] = bi_ee_symmetry_hist_temperature_nsteps_chiral_lag_iter
                
            bi_ee_symmetry_hist_temperature_nsteps[nsteps] = bi_ee_symmetry_hist_temperature_nsteps_chiral_lag 
            
        bi_ee_symmetry_hist_temperature[temperature] = bi_ee_symmetry_hist_temperature_nsteps

    bi_ee_symmetry_hist_density[density] = bi_ee_symmetry_hist_temperature
                    
```
