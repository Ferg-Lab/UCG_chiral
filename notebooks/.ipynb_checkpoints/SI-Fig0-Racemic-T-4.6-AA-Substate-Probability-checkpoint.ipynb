```python
################################################################################################
## SOME HELPFUL FUNCTIONS
################################################################################################

import pandas as pd

def read_lammps_dump_xyz(file_path):

    with open(file_path, 'r') as f:
        lines = f.readlines()
    
    timesteps = {}
    current_timestep = None
    header_line = None
    data_lines = []

    for i, line in enumerate(lines):
        
        line = line.strip()
        
        if line.startswith("ITEM: TIMESTEP"):
            if current_timestep is not None and header_line and data_lines:
                df = pd.DataFrame(data_lines, columns=header_line)
                timesteps[current_timestep] = df
                data_lines = []  # Reset data for the new timestep
            
            current_timestep = int(lines[i + 1].strip())
            header_line = None # reset
        
        elif line.startswith("ITEM: ATOMS"):
            header_line = line.split()[2:] 

        elif line and header_line is not None and not line.startswith("ITEM:"):  
            data_lines.append(list(map(float, line.split())))
            
    if current_timestep is not None and header_line and data_lines:
        df = pd.DataFrame(data_lines, columns=header_line)
        timesteps[current_timestep] = df
    
    return timesteps
```

```python
################################################################################################

#### Read trajs and universe.
import MDAnalysis as mda
import numpy as np

# traj_file_path = '/project2/andrewferguson/sivadasetty/doe/ucg/chiral_tetramer/SystemSize1000tetramers/T-4.6_freqdouble/'
traj_file_path = '/project2/andrewferguson/sivadasetty/doe/ucg/chiral_tetramer/SystemSize1000tetramers/T-' +str(target_temperature)+'_freqdouble_racemic/'


# final trajs
# file_tb_traj_trr = traj_file_path + 'dump_files/traj_reformat_joinfrag_vmd_fixed_forces.lammpstrj'
# file_tb_traj_trr = traj_file_path + 'dump_rerun_files/ucg_traj_reformat_joinfrag_vmd_fixed_forces_nbinary_chirality.lammpstrj'
file_tb_traj_trr = traj_file_path + 'dump_rerun_files/traj_reformat_joinfrag_vmd_fixed_forces_sortedid.lammpstrj'

    
# file_tb_psf = traj_file_path + 'dump_rerun_files/../../../../notebooks/chiral/steepness-gamma-temperature-explore/t_4.6/gamma_0/UCG_tetramer.data'
file_tb_psf = traj_file_path + 'dump_rerun_files/initialconfig.psf'
file_tb_data = traj_file_path + 'initialconfig.dat'

#com_mda_universe = mda.Universe(file_tb_psf, file_tb_traj_trr, format="lammpsdump")
mda_universe = mda.Universe(file_tb_psf, file_tb_traj_trr, format="lammpsdump")
    
# select_atoms = mda_universe.atoms
# transform = mda.transformations.wrap(select_atoms)
# mda_universe.trajectory.add_transformations(transform)
################################################################################################

#### per atom chirality data
# file_path = traj_file_path + 'dump_files/zeta_per_atom_per_frame.tetramer'
file_path = traj_file_path + 'dump_rerun_files/zeta_per_atom_per_frame.tetramer'
zeta_data = read_lammps_dump_xyz(file_path)

###############################################################################################

# #### average chirality data and thermo data
# file_path = traj_file_path + 'dump_files/log_rerun_zeta_per_atom_per_frame.lammps'
# log_data = read_lammps_logfile(file_path)
```

```python
################################################################################################

# compute probability of the chirality of a molecule given its local enantiomeric excess
cutoff = 4.0 #2.0 #2.5 #3.0 #6.0 # 7.5 # 6.0 # 3.0 #2.75

def get_nearest_neighbors_zeta(n_residues, dists, zeta, cutoff):

    # Generate neighbor list (excluding self, i.e., distance > 0)
    nearest_neigbors = {}
    #print(f'n_residues:{n_residues}')
    for i in range(n_residues):
        neighbors = np.where((dists[i] <= cutoff) & (dists[i] > 0))[0]
        neighbors_i = neighbors.tolist()
        neighbor_zeta_i = zeta['c_zatom'].values[neighbors_i]
        neighbor_dists = dists[i, neighbors_i]
        nearest_neigbors[i+1] = {i+1: list(map(lambda x: x + 1, neighbors_i)), 'zeta': neighbor_zeta_i, 
                                 'dists': neighbor_dists, 'zeta_i': zeta['c_zatom'].values[i]}

    #print(f'neighbors:{nearest_neigbors[1]}')
    return nearest_neigbors
    
    
# Collect zeta values of each molecule and its neigbors within a cutoff in each frame
zeta_molecule_its_neighbors = {}
n_frames = mda_universe.trajectory.n_frames
for ts in mda_universe.trajectory[0::1]:

    results_file = traj_file_path + './chiral_results/zeta_molecule_its_neighbors_Sys1000_T_' + str(target_temperature) + '_extended_fr_'+str(int(ts.time))+'_cut'+str(cutoff)+'.pkl'

    if not os.path.exists(results_file):
    
        if int(ts.time) % 10 == 0:
            print(f'{int(ts.time)} of {n_frames} completed.')
        
        # calculate COM of each residue or molecule
        com_all_molecules = mda_universe.select_atoms('all').center_of_mass(compound='residues')
        
        # compute distances between all residues
        com_all_molecules_pairdists = distances.distance_array(com_all_molecules, com_all_molecules, box=mda_universe.dimensions)
        
        # extract chirality of each residue or moleucle
        zeta_frame = zeta_data[int(ts.time)].sort_values(by='mol').drop_duplicates(subset='mol')
        
        n_residues = mda_universe.atoms.n_residues
        
        nearest_neigbors = get_nearest_neighbors_zeta(n_residues, com_all_molecules_pairdists, zeta_frame, cutoff)
        
        #zeta_molecule_its_neighbors[int(ts.time)] = nearest_neigbors
        
        # Open the file in binary read mode and load the dictionary
        with open(results_file, 'wb') as file:
            pickle.dump(nearest_neigbors, file)

    else:
        
        print(f' Reading {int(ts.time)} of {n_frames}.')
        with open(results_file, 'rb') as file:

            zeta_molecule_its_neighbors[int(ts.time)] = pickle.load(file)


```

```python
################################################################################################

def local_proximity_function(r, rth):
    w = 0.5 * (1.0 - np.tanh((r - rth) / (0.1 * rth)))
    return w
    
## load data
ri_zeta_values = []
ri_neigh_zeta_values = []
ri_neigh_dist_values = []
ri_neigh_zeta_values_len = []
ri_neigh_zeta_values_first_nneighs = []
ri_neigh_zeta_values_first_nneighs_dists = []

time_stamp = []

rth = 4.0 # rth: parameter for proximity function.
chirality_cutoff = 0.0 #0.3 # 0.0
nneighs=6

for ts, zeta_data_local in zeta_molecule_its_neighbors.items():

    if int(ts) % 10 == 0: 
        print(f'{int(ts)} completed.')

    for ri in range(1,len(zeta_data_local)+1):
        zeta_molecule_its_neighbors_zeta_vals = zeta_data_local[ri]['zeta']
        zeta_molecule_its_neighbors_zeta_i = zeta_data_local[ri]['zeta_i']
        dists_i = zeta_data_local[ri]['dists']
        sorted_zeta_molecule_its_neighbors_zeta_vals = zeta_molecule_its_neighbors_zeta_vals[np.argsort(dists_i)]

        # Mask for values where |zeta| > 0.3
        #mask = np.abs(zeta_molecule_its_neighbors_zeta_vals) > 0.3 # FOR MASKING
        neighbor_chirality = np.where(sorted_zeta_molecule_its_neighbors_zeta_vals < chirality_cutoff, 1.0, 0.0) # FOR BINARY
        center_chirality = np.where(zeta_molecule_its_neighbors_zeta_i < chirality_cutoff, 1.0, 0.0)

        # neighbor_chirality = [1 if val < chirality_cutoff else 0 if val > chirality_cutoff else None for val in sorted_zeta_molecule_its_neighbors_zeta_vals]
        # neighbor_chirality = [val for val in neighbor_chirality if val is not None]

        # if zeta_molecule_its_neighbors_zeta_i > chirality_cutoff:
        #     center_chirality = 0
        # elif zeta_molecule_its_neighbors_zeta_i < chirality_cutoff:
        #     center_chirality = 1 
        # else:
        #     center_chirality = None 

        neighbor_chirality_len = len(neighbor_chirality)
        
        #filtered_dists = dists_i[mask] # ADD MASK IF NEEDED
        #filtered_signs = signs[mask] # ADD MASK IF NEEDED
        #filtered_zeta = zeta_molecule_its_neighbors_zeta_vals[mask]
        #proximities = local_proximity_function(filtered_dists, rth)

        # SCALED WITH PROXIMITY FUNCTION
        #numerator_l_minus_d_total = np.sum(proximities * filtered_signs) # L-D (indirect; with proximity)
        #numerator_l_minus_d_total = np.sum(proximities * (-1*filtered_zeta)) # SUM OF CHIRALITIES

        ri_zeta_values.append(center_chirality)
        ri_neigh_zeta_values.append(neighbor_chirality)
        ri_neigh_dist_values.append(dists_i[np.argsort(dists_i)])
        ri_neigh_zeta_values_len.append(neighbor_chirality_len)
        ri_neigh_zeta_values_first_nneighs.append(neighbor_chirality[:nneighs])
        ri_neigh_zeta_values_first_nneighs_dists.append(dists_i[np.argsort(dists_i)][:nneighs])
        time_stamp.append(int(ts))
```

```python

ri_centered_neigh_chirality_df = pd.DataFrame({'t': time_stamp, 'c': ri_zeta_values, 'n': ri_neigh_zeta_values,
                                               'd': ri_neigh_dist_values,
                                               'l': ri_neigh_zeta_values_len, 'cn': ri_neigh_zeta_values_first_nneighs,
                                               'dn': ri_neigh_zeta_values_first_nneighs_dists})
```

```python
################################################################################################

# Fit open probability to a function

import json
from scipy.optimize import curve_fit

# def sigmoid_function(x, a, b, c):
# def sigmoid_function(x, a, b):
def sigmoid_function(x, a):
    # return 1.0 / (1 + np.exp(-1*(a * (x - b))))
    # return c / (1 + np.exp(-1*(a * (x - b))))
    return 1.0 / (1 + np.exp(-1*(a * (x - 0.5))))
    

# filtered_df = grouped_ri_centered_neigh_chriality_df_3[(grouped_ri_centered_neigh_chriality_df_3['cn_3_mean'] != -1) ]
# filtered_df = grouped_ri_centered_neigh_chriality_df_4[(grouped_ri_centered_neigh_chriality_df_4['cn_4_mean'] != -1) ]
# filtered_df = grouped_ri_centered_neigh_chriality_df_5[(grouped_ri_centered_neigh_chriality_df_5['cn_5_mean'] != -1) ]
# filtered_df = grouped_ri_centered_neigh_chriality_df_6[(grouped_ri_centered_neigh_chriality_df_6['cn_6_mean'] != -1) ]
# filtered_df = grouped_ri_centered_neigh_chriality_df_8[(grouped_ri_centered_neigh_chriality_df_8['cn_8_mean'] != -1) ]
filtered_df = grouped_ri_centered_neigh_chriality_df_12[(grouped_ri_centered_neigh_chriality_df_12['cn_12_mean'] != -1) ]
# filtered_df = grouped_ri_centered_neigh_chriality_df_15[(grouped_ri_centered_neigh_chriality_df_15['cn_15_mean'] != -1) ]


# x_data = filtered_df['cn_3_mean']
# x_data = filtered_df['cn_4_mean']
# x_data = filtered_df['cn_5_mean']
# x_data = filtered_df['cn_6_mean']
# x_data = filtered_df['cn_8_mean']
x_data = filtered_df['cn_12_mean']
# x_data = filtered_df['cn_15_mean']





y_data = filtered_df['p_c']

################################################################################################
# Sigmoid fit
# initial_guess = [0.5,1.,1.]
# initial_guess = [0.5,1.]
initial_guess = [0.5]

prob_params, prob_covariance = curve_fit(sigmoid_function, x_data, y_data, p0=initial_guess, maxfev=10000)
# print("Fitted Parameters (a, b, c):", prob_params)
print("Fitted Parameters (a):", prob_params)


# model_info = {
#     "model_type": "tanh",
#     "model_degree": len(prob_params) - 1,
#     "parameters": prob_params.tolist()
# }

# with open(traj_file_path + './chiral_results/sig_model_state_probability_Sys1000_T_4.6_extended.json', 'w') as f:
#     json.dump(model_info, f)

x_fit = np.linspace(min(x_data), max(x_data), 1000)
y_fit = sigmoid_function(x_fit, *prob_params)

fit_a = np.round(prob_params[0],3)
fit_b = 0.5 #np.round(prob_params[1],3) #0.0 #np.round(prob_params[1],3)
fit_c = 1.0 #np.round(prob_params[2],3)

label=r'$p_{i}$ = $\frac{'+str(fit_c)+'}{1+exp(-('+str(fit_a)+'(\overline{p}_{neigh,i}-'+str(fit_b)+')))}$'
################################################################################################

fig = plt.figure(figsize=(7.5, 3))
plt.subplots_adjust(left=0.1, right=0.9, bottom=0.1, top=0.9, wspace=0.25, hspace=0.5)

grid = GridSpec(1, 1)

ax1 = plt.subplot(grid[0, 0])

# ax1.plot(x_data, y_data, 'gray', ls=':', lw=4)
ax1.scatter(x_data, y_data, color='black', marker='o', s=50)
ax1.plot(x_fit, y_fit, color='black', lw=2, alpha=1.0, label=label)
# ax1.legend(frameon=False, fontsize=20, handlelength=0, bbox_to_anchor=(.95, -0.2))

ax1.yaxis.set_major_locator(MultipleLocator(0.1))
ax1.xaxis.set_major_locator(MultipleLocator(0.2))
ax1.yaxis.set_minor_locator(AutoMinorLocator(n=5))
ax1.xaxis.set_minor_locator(AutoMinorLocator(n=5))
ax1.tick_params(which='major', length=8)
ax1.tick_params(which='minor', length=4)

ax1.set_xlabel(r'$\overline{p}_{neigh,i}$(L)')
ax1.set_ylabel(r'$p_{i}$(L)')

ax12 = ax1.twinx()
ax12.scatter(x_data, 1-y_data, color='gray', marker='o', s=50)
ax12.plot(x_fit, 1-y_fit, color='gray', lw=2, alpha=1.0, label=label)
ax12.tick_params('y', colors='gray')

ax12.yaxis.set_major_locator(MultipleLocator(0.1))
ax12.yaxis.set_minor_locator(AutoMinorLocator(n=5))
ax12.tick_params(which='major', length=8)
ax12.tick_params(which='minor', length=4)

ax12.set_ylabel(r'$p_{i}$(D)', color='gray')

ax1.grid()
```
