```python

rdf_pot_ll_iter = []
rdf_pot_dd_iter = []
rdf_pot_ld_iter = []

for i in range(maxiter):

    diff_t1 =  np.sum(np.abs(bi_ucg_RDF_max_6[i]['t1']['rdf']-symm_aa_RDF_max_6['t1']['rdf']))
    diff_t1 /= np.sum(np.abs(bi_ucg_RDF_max_6[i]['t1']['rdf'])+np.abs(symm_aa_RDF_max_6['t1']['rdf']))

    diff_t2 =  np.sum(np.abs(bi_ucg_RDF_max_6[i]['t2']['rdf']-symm_aa_RDF_max_6['t2']['rdf']))
    diff_t2 /= np.sum(np.abs(bi_ucg_RDF_max_6[i]['t2']['rdf'])+np.abs(symm_aa_RDF_max_6['t2']['rdf']))

    diff_t12 =  np.sum(np.abs(bi_ucg_RDF_max_6[i]['t12']['rdf']-symm_aa_RDF_max_6['t12']['rdf']))
    diff_t12 /= np.sum(np.abs(bi_ucg_RDF_max_6[i]['t12']['rdf'])+np.abs(symm_aa_RDF_max_6['t12']['rdf']))

    rdf_pot_ll_iter.append(1-diff_t1)
    rdf_pot_dd_iter.append(1-diff_t2)
    rdf_pot_ld_iter.append(1-diff_t12)    
```
