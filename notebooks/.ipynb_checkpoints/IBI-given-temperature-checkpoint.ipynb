```python

rdf_pot_ll_iter = []
rdf_pot_dd_iter = []
rdf_pot_ld_iter = []

for i in range(maxiter):

    diff_t1 =  np.sum(np.abs(bi_ucg_RDF_max_6[i]['t1']['rdf']-aa_RDF_max_6['t1']['rdf']))
    diff_t1 /= np.sum(np.abs(bi_ucg_RDF_max_6[i]['t1']['rdf'])+np.abs(aa_RDF_max_6['t1']['rdf']))

    diff_t2 =  np.sum(np.abs(bi_ucg_RDF_max_6[i]['t2']['rdf']-aa_RDF_max_6['t2']['rdf']))
    diff_t2 /= np.sum(np.abs(bi_ucg_RDF_max_6[i]['t2']['rdf'])+np.abs(aa_RDF_max_6['t2']['rdf']))

    diff_t12 =  np.sum(np.abs(bi_ucg_RDF_max_6[i]['t12']['rdf']-aa_RDF_max_6['t12']['rdf']))
    diff_t12 /= np.sum(np.abs(bi_ucg_RDF_max_6[i]['t12']['rdf'])+np.abs(aa_RDF_max_6['t12']['rdf']))

    rdf_pot_ll_iter.append(1-diff_t1)
    rdf_pot_dd_iter.append(1-diff_t2)
    rdf_pot_ld_iter.append(1-diff_t12)    
```

```python

fig = plt.figure(figsize=(5, 3))
plt.subplots_adjust(left=0.1, right=0.9, bottom=0.1, top=0.9, wspace=0.35, hspace=0.2)
grid = GridSpec(1, 1)
ax1 = plt.subplot(grid[0, 0])
ax1.plot(np.arange(len(rdf_pot_ll_iter)), np.array(rdf_pot_ll_iter), color='blue')
ax1.plot(np.arange(len(rdf_pot_dd_iter)), np.array(rdf_pot_dd_iter), color='green')
ax1.plot(np.arange(len(rdf_pot_ld_iter)), np.array(rdf_pot_ld_iter), color='red')
print(np.argmin(rdf_pot_ld_iter))
print(np.argmax(rdf_pot_ld_iter[1:]))

# ax1.set_ylim(0,1)
# ax1.set_xlim(0,30)
```
