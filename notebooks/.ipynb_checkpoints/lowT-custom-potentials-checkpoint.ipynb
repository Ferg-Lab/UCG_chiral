```python

net_path = '/project2/andrewferguson/sivadasetty/doe/ucg/notebooks/chiral/manuscript/simulations/low-t/'

fig = plt.figure(figsize=(10, 12))
plt.subplots_adjust(left=0.1, right=0.9, bottom=0.1, top=0.9, wspace=0.3, hspace=0.2)
grid = GridSpec(3, 2)
n = 1 # number of colors
colors = plt.cm.rainbow(np.linspace(0, 1, n))  # or use 'plasma', 'inferno', 'tab10', etc.
colors = ['blue', 'black', 'red']

targetpath = 'manual'

change_pot_ll_iter = []
change_pot_dd_iter = []
change_pot_ld_iter = []

for k, targetpath in enumerate([2.0]):

    saved_pot_ll = pd.read_csv(net_path + 't_'+str(targetpath)+ 'gamma_0/prior_model' + '/prior_model/Pair_L-L.table', skiprows=4, 
                               sep='\s+', names=['r', 'u', 'f'])
    saved_pot_dd = pd.read_csv(net_path + 't_'+str(targetpath)+ 'gamma_0/prior_model' + '/prior_model/Pair_D-D.table', skiprows=4, 
                               sep='\s+', names=['r', 'u', 'f'])
    saved_pot_ld = pd.read_csv(net_path + 't_'+str(targetpath)+ 'gamma_0/prior_model' + '/prior_model/Pair_L-D.table', skiprows=4, 
                               sep='\s+', names=['r', 'u', 'f'])
            
    xj = [saved_pot_ll.r, saved_pot_dd.r, saved_pot_ld.r]
    yj = [saved_pot_ll.u, saved_pot_dd.u, saved_pot_ld.u] 

    for j in [0,2]:
        ax1 = plt.subplot(grid[k, 0])
        ax1.plot(xj[j], yj[j], color=colors[j], lw=2)
        ax1.scatter(xj[j], yj[j], color=colors[j], s=10)
        
    
        ax1.set_xlabel(r"$r$")
        ax1.set_ylabel("U($r$)")
    
        ax1.set_ylim(-3,6)
        
        ax1.xaxis.set_major_locator(MultipleLocator(1))
        ax1.xaxis.set_minor_locator(AutoMinorLocator(n=5))
        ax1.yaxis.set_major_locator(MultipleLocator(1.))
        ax1.yaxis.set_minor_locator(AutoMinorLocator(n=5))
    
    ax1.grid()

    xj = [saved_pot_ll.r, saved_pot_dd.r, saved_pot_ld.r]
    yj = [saved_pot_ll.f, saved_pot_dd.f, saved_pot_ld.f]

    labels=['HO','','HE']
    for j in [0,2]:
        ax2 = plt.subplot(grid[k, 1])
        ax2.scatter(xj[j], yj[j], color=colors[j], s=10)
        ax2.plot(xj[j], yj[j], color=colors[j], lw=2, label=labels[j])
        
        ax2.set_ylim(-3,6)
        
        ax2.set_xlabel(r"$r$")
        ax2.set_ylabel("F($r$)")
    
        ax2.xaxis.set_major_locator(MultipleLocator(1))
        ax2.yaxis.set_major_locator(MultipleLocator(1.)) #0.5))
        ax2.xaxis.set_minor_locator(AutoMinorLocator(n=5))
        ax2.yaxis.set_minor_locator(AutoMinorLocator(n=5))
        #ax2.axhline(y=0)

    ax2.grid()
    ax2.legend(frameon=False)
```

## Compare chiral counts

```python

fig = plt.figure(figsize=(15, 3))
plt.subplots_adjust(left=0.1, right=0.9, bottom=0.1, top=0.9, wspace=0.35, hspace=0.2)
grid = GridSpec(1, 3)


ax1 = plt.subplot(grid[0, 0])
ax1.plot(np.arange(len(chiralcount_max_6['l'])), chiralcount_max_6['l'].values(), color='blue')
ax1.plot(np.arange(len(chiralcount_max_6['d'])), chiralcount_max_6['d'].values(), color='red')

    
l_hist, l_bin_edges = np.histogram(np.array(list(chiralcount_max_6['l'].values())), bins=50, density=True)
d_hist, d_bin_edges = np.histogram(np.array(list(chiralcount_max_6['d'].values())), bins=50, density=True)

l_bin_centers = 0.5 * (l_bin_edges[:-1] + l_bin_edges[1:])
d_bin_centers = 0.5 * (d_bin_edges[:-1] + d_bin_edges[1:])

aa_chiral_hist_temperature = {'l_hist': l_hist, 'l_bin_edges': l_bin_edges, 'l_bin_centers': l_bin_centers,
                              'd_hist': d_hist, 'd_bin_edges': d_bin_edges, 'd_bin_centers': d_bin_centers}

numerator = np.array(list(chiralcount_max_6['l'].values()))-np.array(list(chiralcount_max_6['d'].values()))
denominator = np.array(list(chiralcount_max_6['l'].values()))+np.array(list(chiralcount_max_6['d'].values()))
aa_ee = np.array(numerator)/np.array(denominator)
    

ax1.set_xlabel(r"$Frame$")
ax1.set_ylabel("Count")
    
ax1.xaxis.set_major_locator(MultipleLocator(4000))
ax1.yaxis.set_major_locator(MultipleLocator(200))
ax1.xaxis.set_minor_locator(AutoMinorLocator(n=5))
ax1.yaxis.set_minor_locator(AutoMinorLocator(n=5))
ax1.grid()

ax2 = plt.subplot(grid[0, 1])

ax2.plot(np.arange(len(manual_ucg_chiralcount_max_6['l'])), manual_ucg_chiralcount_max_6['l'].values(), color='blue')
ax2.plot(np.arange(len(manual_ucg_chiralcount_max_6['d'])), manual_ucg_chiralcount_max_6['d'].values(), color='red')

l_hist, l_bin_edges = np.histogram(np.array(list(manual_ucg_chiralcount_max_6['l'].values())), bins=50, density=True)
d_hist, d_bin_edges = np.histogram(np.array(list(manual_ucg_chiralcount_max_6['d'].values())), bins=50, density=True)

l_bin_centers = 0.5 * (l_bin_edges[:-1] + l_bin_edges[1:])
d_bin_centers = 0.5 * (d_bin_edges[:-1] + d_bin_edges[1:])

manual_ucg_chiral_hist_temperature = {'l_hist': l_hist, 'l_bin_edges': l_bin_edges, 'l_bin_centers': l_bin_centers,
                                      'd_hist': d_hist, 'd_bin_edges': d_bin_edges, 'd_bin_centers': d_bin_centers}

numerator = (np.array(list(manual_ucg_chiralcount_max_6['l'].values()))-np.array(list(manual_ucg_chiralcount_max_6['d'].values())))
denominator = (np.array(list(manual_ucg_chiralcount_max_6['l'].values()))+np.array(list(manual_ucg_chiralcount_max_6['d'].values())))
manual_ee = np.array(numerator)/np.array(denominator)

ax2.set_xlabel(r"$Frame$")
ax2.set_ylabel("Count")
        
ax2.xaxis.set_major_locator(MultipleLocator(500))
ax2.yaxis.set_major_locator(MultipleLocator(200))
ax2.xaxis.set_minor_locator(AutoMinorLocator(n=5))
ax2.yaxis.set_minor_locator(AutoMinorLocator(n=5))
ax2.grid()

ax3 = plt.subplot(grid[0, 2])

ax3.plot(np.arange(len(bi_ucg_chiralcount_max_6['l'])), bi_ucg_chiralcount_max_6['l'].values(), color='blue')
ax3.plot(np.arange(len(bi_ucg_chiralcount_max_6['d'])), bi_ucg_chiralcount_max_6['d'].values(), color='red')

l_hist, l_bin_edges = np.histogram(np.array(list(bi_ucg_chiralcount_max_6['l'].values())), bins=50, density=True)
d_hist, d_bin_edges = np.histogram(np.array(list(bi_ucg_chiralcount_max_6['d'].values())), bins=50, density=True)

l_bin_centers = 0.5 * (l_bin_edges[:-1] + l_bin_edges[1:])
d_bin_centers = 0.5 * (d_bin_edges[:-1] + d_bin_edges[1:])

bi_ucg_chiral_hist_temperature = {'l_hist': l_hist, 'l_bin_edges': l_bin_edges, 'l_bin_centers': l_bin_centers,
                                  'd_hist': d_hist, 'd_bin_edges': d_bin_edges, 'd_bin_centers': d_bin_centers}

numerator = (np.array(list(bi_ucg_chiralcount_max_6['l'].values()))-np.array(list(bi_ucg_chiralcount_max_6['d'].values())))
denominator = (np.array(list(bi_ucg_chiralcount_max_6['l'].values()))+np.array(list(bi_ucg_chiralcount_max_6['d'].values())))
bi_ee = np.array(numerator)/np.array(denominator)

ax3.set_xlabel(r"$Frame$")
ax3.set_ylabel("Count")
        
ax3.xaxis.set_major_locator(MultipleLocator(10000))
ax3.yaxis.set_major_locator(MultipleLocator(200))
ax3.xaxis.set_minor_locator(AutoMinorLocator(n=5))
ax3.yaxis.set_minor_locator(AutoMinorLocator(n=5))
ax3.grid()
```
