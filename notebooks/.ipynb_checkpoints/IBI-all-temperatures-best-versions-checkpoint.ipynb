## UCG potentials and RDF

```python

results_path = '/project2/andrewferguson/sivadasetty/doe/ucg/notebooks/chiral/manuscript/simulations/'

explore_temperature_list = [2.0, 2.5, 2.7, 2.9, 3.0, 3.2, 3.5, 3.8, 4.0, 4.2, 4.4, 4.6, 4.8, 5.0]
maxiter=60
potentials_temperature = {}
bi_ucg_RDF_max_6_temperature = {}
rdf_match_temperature = {}
symm_aa_RDF_max_6_temperature = {}

for temperature in explore_temperature_list:

    ################################################################################################
    # Read AA RDF [See ../scripts/run_rdf.py  for calculation details]
    ################################################################################################
    aa_traj_file_path = '/project2/andrewferguson/sivadasetty/doe/ucg/chiral_tetramer/SystemSize1000tetramers/T-'+str(temperature)+'_freqdouble_racemic/'
        
    symm_rdf_results_file_max_6 = aa_traj_file_path +  'chiral_results/symmetry_rdf_halfbox_maxbound_8_nbinary.pkl'
    with open(symm_rdf_results_file_max_6, 'rb') as file:
        symm_aa_RDF_max_6 = pickle.load(file)

    symm_aa_RDF_max_6_temperature[temperature] = symm_aa_RDF_max_6
    
    if temperature <= 3.0:
        ibi_target_folder = results_path + 'ibi/low-t/t_'+str(temperature)
    else:
        ibi_target_folder = results_path + 'ibi/high-t/t_'+str(temperature)
            
    # Read potentials
    potentials_iter = {}
    for i in range(maxiter):
        
        ibi_saved_pot_ll = pd.read_csv(ibi_target_folder + '/iter_'+str(i)+'/Pair_L-L.table', skiprows=4, 
                                       sep='\s+', names=['r', 'u', 'f'])
        ibi_saved_pot_dd = pd.read_csv(ibi_target_folder + '/iter_'+str(i)+'/Pair_D-D.table', skiprows=4, 
                                       sep='\s+', names=['r', 'u', 'f'])
        ibi_saved_pot_ld = pd.read_csv(ibi_target_folder + '/iter_'+str(i)+'/Pair_L-D.table', skiprows=4, 
                                       sep='\s+', names=['r', 'u', 'f'])
    
        potentials_iter[i] = {'ll': ibi_saved_pot_ll, 'dd': ibi_saved_pot_dd, 'ld': ibi_saved_pot_ld}

    potentials_temperature[temperature] = potentials_iter

    ################################################################################################
    # Read UCG RDF [See ../scripts/run_rdf.py  for calculation details]
    ################################################################################################
    
    bi_ucg_RDF_max_6 = {}
    for i in range(maxiter):
   
        rdf_results_file_max_6 = ibi_target_folder + '/iter_' + str(i) + '/' + 'symmetry_rdf_halfbox_maxbound_8_nbinary.pkl'
        with open(rdf_results_file_max_6, 'rb') as file:
            bi_ucg_RDF_max_6[i] = pickle.load(file)
    

    bi_ucg_RDF_max_6_temperature[temperature] = bi_ucg_RDF_max_6



    rdf_pot_ll_iter = []
    rdf_pot_dd_iter = []
    rdf_pot_ld_iter = []
    
    for i in range(maxiter):
    
        diff_t1 =  np.sum(np.abs(bi_ucg_RDF_max_6[i]['t1']['rdf']-symm_aa_RDF_max_6_temperature[temperature]['t1']['rdf']))
        diff_t1 /= np.sum(np.abs(bi_ucg_RDF_max_6[i]['t1']['rdf'])+np.abs(symm_aa_RDF_max_6_temperature[temperature]['t1']['rdf']))
    
        diff_t2 =  np.sum(np.abs(bi_ucg_RDF_max_6[i]['t2']['rdf']-symm_aa_RDF_max_6_temperature[temperature]['t2']['rdf']))
        diff_t2 /= np.sum(np.abs(bi_ucg_RDF_max_6[i]['t2']['rdf'])+np.abs(symm_aa_RDF_max_6_temperature[temperature]['t2']['rdf']))
    
        diff_t12 =  np.sum(np.abs(bi_ucg_RDF_max_6[i]['t12']['rdf']-symm_aa_RDF_max_6_temperature[temperature]['t12']['rdf']))
        diff_t12 /= np.sum(np.abs(bi_ucg_RDF_max_6[i]['t12']['rdf'])+np.abs(symm_aa_RDF_max_6_temperature[temperature]['t12']['rdf']))
    
        rdf_pot_ll_iter.append(1-diff_t1)
        rdf_pot_dd_iter.append(1-diff_t2)
        rdf_pot_ld_iter.append(1-diff_t12)    

    rdf_match_temperature[temperature] = {'ll': rdf_pot_ll_iter, 'dd': rdf_pot_dd_iter, 'ld': rdf_pot_ld_iter}
```

```python
target_temperature = 5.0
```

```python

import matplotlib.cm as cm
colors = cm.viridis(np.linspace(0, 1, maxiter))

fig = plt.figure(figsize=(10, 10))
grid = GridSpec(4, 2)
plt.subplots_adjust(left=0.1, right=0.9, bottom=0.1, top=0.9, wspace=0.35, hspace=0.45)

axes = []
for i in range(4):
    for j in range(2):
        if i == 3 and j == 1:
            continue
        axes.append(plt.subplot(grid[i,j]))

for i, ax in enumerate(axes[:-1]):
    ax.grid()
    ax.xaxis.set_major_locator(MultipleLocator(1))
    ax.xaxis.set_minor_locator(AutoMinorLocator(n=5))
    ax.yaxis.set_major_locator(MultipleLocator(5))

    if i > 3:
        ax.yaxis.set_major_locator(MultipleLocator(.5))


for i in [0, 2]:
    axes[i].set_xlabel(r"$r$")
    axes[i].set_ylabel("U($r$)")
    axes[i].set_ylim(-5,20)

for i in [1, 3]:
    axes[i].set_xlabel(r"$r$")
    axes[i].set_ylabel("F($r$)")
    axes[i].set_ylim(-5,20)

for i in [4, 5]:
    axes[i].set_xlabel(r"$r$")
    axes[i].set_ylabel("g($r$)")

axes[6].set_xlabel('iter')
axes[6].set_ylabel(r'f$_{fit}$')


best_match_temperature = {}
for temperature in [target_temperature]:

    for i in range(maxiter):
        
        axes[0].plot(potentials_temperature[temperature][i]['ll'].r, potentials_temperature[temperature][i]['ll'].u, color=colors[i], lw=2)     
        axes[1].plot(potentials_temperature[temperature][i]['ll'].r, potentials_temperature[temperature][i]['ll'].f, color=colors[i], lw=2)
        axes[2].plot(potentials_temperature[temperature][i]['ld'].r, potentials_temperature[temperature][i]['ld'].u, color=colors[i], lw=2)
        axes[3].plot(potentials_temperature[temperature][i]['ld'].r, potentials_temperature[temperature][i]['ld'].f, color=colors[i], lw=2)

        axes[4].plot(bi_ucg_RDF_max_6_temperature[temperature][i]['t1']['bin_centers'], bi_ucg_RDF_max_6_temperature[temperature][i]['t1']['rdf']*2, color=colors[i], lw=2)
        axes[5].plot(bi_ucg_RDF_max_6_temperature[temperature][i]['t12']['bin_centers'], bi_ucg_RDF_max_6_temperature[temperature][i]['t12']['rdf'], color=colors[i], lw=2)

    axes[4].plot(symm_aa_RDF_max_6_temperature[temperature]['t1']['bin_centers'], symm_aa_RDF_max_6_temperature[temperature]['t1']['rdf']*2, color='red', lw=2)
    axes[5].plot(symm_aa_RDF_max_6_temperature[temperature]['t12']['bin_centers'], symm_aa_RDF_max_6_temperature[temperature]['t12']['rdf'], color='red', lw=2)


    axes[6].plot(np.arange(len(rdf_match_temperature[temperature]['ll'])), np.array(rdf_match_temperature[temperature]['ll']), color='gray', lw=2)
    axes[6].plot(np.arange(len(rdf_match_temperature[temperature]['ll'])), np.array(rdf_match_temperature[temperature]['ld']), color='black', lw=2)

    # print(np.argmax(np.array(rdf_match_temperature[temperature]['ll'])))
    # print(np.argmax(np.array(rdf_match_temperature[temperature]['ld'])))

    best_match_temperature[temperature] = {'ll': np.argmax(np.array(rdf_match_temperature[temperature]['ll'])),
                                           'ld': np.argmax(np.array(rdf_match_temperature[temperature]['ld']))}

```
