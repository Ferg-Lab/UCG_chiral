```python
def acf_direct_scalar(e, delta_t=1.0, max_lag=None):
    """
    Direct implementation of:
        C(Δt) = <e(t+Δt) e(t)> / <e(t) e(t)>
    for a scalar time series e(t), with no mean removal.

    Parameters
    ----------
    e : array_like, shape (N,)
        Scalar time series values sampled uniformly in time.
    delta_t : float, optional
        Sampling interval between consecutive points (default 1.0).
    max_lag : int or None
        Maximum lag (in frames). If None, computes all lags 0..N-1.

    Returns
    -------
    t : ndarray, shape (L,)
        Lag times Δt_k = k * delta_t.
    C : ndarray, shape (L,)
        Normalized autocorrelation values at those lags.
    """
    e = np.asarray(e, dtype=float)
    N = e.size
    if max_lag is None:
        max_lag = N - 1
    max_lag = int(min(max_lag, N - 1))

    # Denominator: <e(t) e(t)> over all t (independent of lag)
    denom = np.mean(e * e)
    if denom == 0.0:
        raise ValueError("Denominator <e^2> is zero; cannot normalize.")

    C = np.empty(max_lag + 1, dtype=float)

    # Lag k = 0..max_lag
    for k in range(max_lag + 1):
        # Numerator average over available pairs: <e_{i+k} e_i> with i=0..N-k-1
        num = np.dot(e[:N - k], e[k:]) / (N - k)
        C[k] = num / denom

    t = np.arange(max_lag + 1, dtype=float) * delta_t
    return t, C


def fit_tau_simple(t, C, tmin=None, tmax=None):
    """
    Fit C(Δt) = exp(-Δt / τ_in) to find τ_in.

    Parameters
    ----------
    t : array_like
        Time (Δt) values.
    C : array_like
        Correlation values (Cee(Δt)), should be positive and normalized.
    tmin, tmax : float, optional
        Time window [tmin, tmax] for fitting.

    Returns
    -------
    tau_in : float
        Estimated decay constant τ_in.
    """
    t = np.asarray(t)
    C = np.asarray(C)

    # Select fit window
    mask = (C > 0)
    if tmin is not None:
        mask &= (t >= tmin)
    if tmax is not None:
        mask &= (t <= tmax)

    t_fit = t[mask]
    C_fit = C[mask]

    if len(t_fit) < 2:
        raise ValueError("Not enough valid points in the fitting window.")

    # Linearize: ln(C) = -(1/τ) * t
    y = np.log(C_fit)
    m, b = np.polyfit(t_fit, y, 1)  # slope m ≈ -1/τ_in

    tau_in = -1.0 / m
    return tau_in

```
