## Read AA chiral counts

```python
all_temperatures = [3.8, 4.5, 4.8]
```

```python
################################################################################################
# Read AA chiral count [See ../scripts/run_rdf.py  for calculation details]
################################################################################################
chiral_count_density = {}

for density in [0.07, 0.09, 0.11]:

    if density == 0.11:
        outpath = '/project2/andrewferguson/sivadasetty/doe/ucg/chiral_tetramer/SystemSize1000tetramers/'
    else:
        outpath = '/project2/andrewferguson/sivadasetty/doe/ucg/chiral_tetramer/SystemSize1000tetramers_'+ str(density) + '/'
    
    chiralcount_max_6 = {}
    for temperature in all_temperatures:

        
        aa_traj_file_path = outpath + '/T-'+str(temperature)+'_freqdouble_racemic/'
        
        chiral_results_file_max_6 = aa_traj_file_path +  'chiral_results/count_ld.pkl'
        
        with open(chiral_results_file_max_6, 'rb') as file:
            chiralcount_max_6[temperature] = pickle.load(file)

    chiral_count_density[density] = chiralcount_max_6
```

## PLOT

```python
fig = plt.figure(figsize=(18, 15))
grid = GridSpec(4, 4)
plt.subplots_adjust(left=0.1, right=0.9, bottom=0.1, top=0.9, wspace=0.45, hspace=0.3)

for j, density in enumerate([0.07, 0.09, 0.11]):
    for k, temperature in enumerate(all_temperatures):
        
        ax11 = plt.subplot(grid[j, k])
        ax11.grid()
        ax11.set_xlabel(r'Time')
        ax11.set_ylabel(r'Count($\rho$:'+str(density)+',T:'+str(temperature)+')')
    
        ax11.scatter(list(chiral_count_density[density][temperature]['l'].keys()), list(chiral_count_density[density][temperature]['l'].values()), color='blue')#, lw=2)
        ax11.plot(list(chiral_count_density[density][temperature]['l'].keys()), list(chiral_count_density[density][temperature]['l'].values()), color='blue')#, lw=2)
    
        ax11.scatter(list(chiral_count_density[density][temperature]['d'].keys()), list(chiral_count_density[density][temperature]['d'].values()), color='red')#, lw=2)
        ax11.plot(list(chiral_count_density[density][temperature]['d'].keys()), list(chiral_count_density[density][temperature]['d'].values()), color='red')#, lw=2)

        # ax11.set_xlim(0,100)

    # ax12 = plt.subplot(grid[i, 1])
    # ax12.grid()
    # # ax12.set_xlim(0,maxiter)
    # # ax12.set_ylim(0.95, 1)
    # ax12.set_ylabel(r'f$_{fit}$')
    # ax12.set_xlabel('Iter')
    # ax12.scatter(list(fit_rdf_unlike_temperature[temperature][nstep][chiral_lag].keys()), list(fit_rdf_unlike_temperature[temperature][nstep][chiral_lag].values()), color='black')#, lw=2)
    # ax12.plot(list(fit_rdf_unlike_temperature[temperature][nstep][chiral_lag].keys()), list(fit_rdf_unlike_temperature[temperature][nstep][chiral_lag].values()), color='black')#, lw=2)
    
    # ax13 = plt.subplot(grid[i, 2])
    # ax13.grid()
    # # ax13.set_xlim(0,maxiter)
    # # ax13.set_ylim(0., 0.5)
    # ax13.set_ylabel(r'KL(EE$_{AA}$||EE$_{UCG}$)')
    # ax13.set_xlabel('Iter')
    # ax13.scatter(list(kl_ee_temperature[temperature][nstep][chiral_lag].keys()), list(kl_ee_temperature[temperature][nstep][chiral_lag].values()), color='black')#, lw=2)
    # ax13.plot(list(kl_ee_temperature[temperature][nstep][chiral_lag].keys()), list(kl_ee_temperature[temperature][nstep][chiral_lag].values()), color='black')#, lw=2)

```
