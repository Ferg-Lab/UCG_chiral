```python
U_ld = uld_pot.copy()
U_ll = ull_pot.copy()

first_finite_idx = np.where(np.isfinite(U_ld))[0][1] #[3]
for i in range(first_finite_idx, -1, -1):
    U_ld[i] = 1.25 * U_ld[i + 1]
    
# first_finite_idx = np.where(np.isfinite(U_ll))[0][0] #[3]
first_finite_idx = np.where(np.isfinite(U_ll))[0][3] #[3]

for i in range(first_finite_idx, -1, -1):
    U_ll[i] = 1.15 * U_ll[i + 1]
    
```

```python
import sys
sys.path.append('../simulations/all-t/')
from bspline import *
from write_potentials import *
```

```python
interior = np.r_[
        0.10:1.5:0.25,    # coarse repulsive wall
        1.5:3.:1,     # dense around coordination shells
        3.:7.0:1     # coarse tail to cutoff
    ]

xmin = 0.01
xmax = 7
degree=3 
model_ld = SplineModel(xmin=xmin, xmax=xmax, degree=degree, interior=interior)
# c_ld = model_ld.fit(xld[((xld<1.) | (xld> 3.))], U_ld[( (xld<1.) | (xld> 3.))], lam=1e-2)
c_ld = model_ld.fit(xld[(xld>0.8)], U_ld[(xld>0.8)], lam=1e-2) # T=2.0
# c_ld = model_ld.fit(xld, U_ld, lam=1e-2)

model_ll = SplineModel(xmin=xmin, xmax=xmax, degree=degree, interior=interior)
# c_ll = model_ld.fit(xld[((xld<1.2) | (xld> 3.))], U_ld[( (xld<1.2) | (xld> 3.))], lam=1e-2)
c_ll = model_ll.fit(xll[(xll>0.3)], U_ll[(xll>0.3)], lam=1e-2)
# c_ll = model_ll.fit(xll, U_ll, lam=1e-2)

xeval = np.arange(0.03, 6, 0.01)
U_ld_pred = model_ld.eval(xeval)
U_ll_pred = model_ll.eval(xeval)

# Smoothen the potentials using Equation 7 in [http://dx.doi.org/10.1063/1.4880555]
for n in range(1, len(U_ll_pred)-1):
    U_ll_pred[n] = (U_ll_pred[n-1] + U_ll_pred[n] + U_ll_pred[n+1]) / 3
U_ll_pred[0] = U_ll_pred[0]  # or some other rule
U_ll_pred[-1] = U_ll_pred[-1]
U_ll_pred -= U_ll_pred[-1]# Smoothen the potentials using Equation 7 in [http://dx.doi.org/10.1063/1.4880555]

for n in range(1, len(U_ld_pred)-1):
    U_ld_pred[n] = (U_ld_pred[n-1] + U_ld_pred[n] + U_ld_pred[n+1]) / 3
U_ld_pred[0] = U_ld_pred[0]  # or some other rule
U_ld_pred[-1] = U_ld_pred[-1]
U_ld_pred -= U_ld_pred[-1]

# T=2.0
U_ld_pred = U_ld_pred*3 #8.5 #8.5 #(4.6/2.0)**3
U_ll_pred = U_ll_pred*0.5 #(4.6/2.0)**0

# # T=2.5
# U_ld_pred = U_ld_pred*8.5 #8.5 #(4.6/2.0)**3
# U_ll_pred = U_ll_pred*1 #(4.6/2.0)**0

ull_force = -np.gradient(U_ll_pred,xeval)
ull_force -= ull_force[-1]
uld_force = -np.gradient(U_ld_pred,xeval)
uld_force -= uld_force[-1]

```
