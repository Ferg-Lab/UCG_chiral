```python


explore_temperature_list = [2.7]
for new_t in explore_temperature_list:
    for gamma in [0]:

        #target_folder = results_path + 'ref-t-4.6/potential_BI/'
        target_folder = results_path + 'custom-t/t_'+str(new_t)+'/gamma_0/'
        
        # target_folder = results_path + 'master_4.6/'
        
        saved_pot_ll = pd.read_csv(target_folder + '/prior_model/Pair_L-L.table', skiprows=4, 
                                   sep='\s+', names=['r', 'u', 'f'])
        saved_pot_dd = pd.read_csv(target_folder + '/prior_model/Pair_D-D.table', skiprows=4, 
                                   sep='\s+', names=['r', 'u', 'f'])
        saved_pot_ld = pd.read_csv(target_folder + '/prior_model/Pair_L-D.table', skiprows=4, 
                                   sep='\s+', names=['r', 'u', 'f'])

        # ibi_target_folder = results_path + 'ibi/high-t/ibi-files/iter_0/'
        ibi_target_folder = results_path + 'ibi/low-t/ibi-files/iter_0_t_2.7/'
        
        
        ibi_saved_pot_ll = pd.read_csv(ibi_target_folder + '/Pair_L-L.table', skiprows=4, 
                                   sep='\s+', names=['r', 'u', 'f'])
        ibi_saved_pot_dd = pd.read_csv(ibi_target_folder + '/Pair_D-D.table', skiprows=4, 
                                   sep='\s+', names=['r', 'u', 'f'])
        ibi_saved_pot_ld = pd.read_csv(ibi_target_folder + '/Pair_L-D.table', skiprows=4, 
                                   sep='\s+', names=['r', 'u', 'f'])
        

        fig = plt.figure(figsize=(10, 3))
        
        plt.subplots_adjust(left=0.1, right=0.9, bottom=0.1, top=0.9, wspace=0.35, hspace=0.2)
        
        grid = GridSpec(1, 2)
        
        ax1 = plt.subplot(grid[0, 0])
        
        ax1.plot(saved_pot_ll.r, saved_pot_ll.u, color='blue', lw=2)
        ax1.plot(saved_pot_dd.r, saved_pot_dd.u, color='green', lw=2)
        ax1.plot(saved_pot_ld.r, saved_pot_ld.u, color='red', lw=2)


        ax1.plot(ibi_saved_pot_ll.r, ibi_saved_pot_ll.u, color='blue', lw=4, ls=':')
        ax1.plot(ibi_saved_pot_dd.r, ibi_saved_pot_dd.u, color='green', lw=4, ls=':')
        ax1.plot(ibi_saved_pot_ld.r, ibi_saved_pot_ld.u, color='red', lw=4, ls=':')

        
        
        ax1.set_xlabel(r"$r$")
        ax1.set_ylabel("U($r$)")
        
        ax1.set_ylim(-5,4)
        # ax1.set_xlim(4,7)
        
        ax1.xaxis.set_major_locator(MultipleLocator(1))
        # ax1.yaxis.set_major_locator(MultipleLocator(.5))
        ax1.xaxis.set_minor_locator(AutoMinorLocator(n=5))
        # ax1.yaxis.set_minor_locator(AutoMinorLocator(n=5))
        # ax1.set_ylim(0.5,2)
        # ax1.set_ylim(0.5,2)
        
        
        ax1.grid()
        ax1.legend(frameon=False)
        
        # ax1.xaxis.set_major_locator(MultipleLocator(1))
        # # ax1.yaxis.set_major_locator(MultipleLocator(0.1)) #025))
        # ax1.xaxis.set_minor_locator(AutoMinorLocator(n=5))
        # ax1.yaxis.set_minor_locator(AutoMinorLocator(n=5))
        
        ax2 = plt.subplot(grid[0, 1])
        
        # ax2.scatter(saved_pot_ll.r, saved_pot_ll.f, color='blue', s=5)
        ax2.plot(saved_pot_ll.r, saved_pot_ll.f, color='blue', lw=2)
        # ax2.scatter(saved_pot_dd.r, saved_pot_dd.f, color='green', s=1)
        ax2.plot(saved_pot_dd.r, saved_pot_dd.f, color='green', lw=1)
        # ax2.scatter(saved_pot_ld.r, saved_pot_ld.f, color='red', s=10)
        ax2.plot(saved_pot_ld.r, saved_pot_ld.f, color='red', lw=2)

        # ax2.scatter(ibi_saved_pot_ll.r, ibi_saved_pot_ll.f, color='blue', s=5)
        ax2.plot(ibi_saved_pot_ll.r, ibi_saved_pot_ll.f, color='blue', lw=4, ls=':')
        # ax2.scatter(ibi_saved_pot_dd.r, ibi_saved_pot_dd.f, color='green', s=1)
        ax2.plot(ibi_saved_pot_dd.r, ibi_saved_pot_dd.f, color='green', lw=4, ls=':')
        # ax2.scatter(ibi_saved_pot_ld.r, ibi_saved_pot_ld.f, color='red', s=10)
        ax2.plot(ibi_saved_pot_ld.r, ibi_saved_pot_ld.f, color='red', lw=4, ls=':')
        
        
        ax2.grid()
        ax2.set_ylim(-5,5)
        # ax2.set_xlim(1,7)
        
        ax2.set_xlabel(r"$r$")
        ax2.set_ylabel("F($r$)")
        
        
        ax2.xaxis.set_major_locator(MultipleLocator(1))
        # ax2.yaxis.set_major_locator(MultipleLocator(1.)) #0.5))
        ax2.xaxis.set_minor_locator(AutoMinorLocator(n=5))
        ax2.yaxis.set_minor_locator(AutoMinorLocator(n=5))
        

```
