```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
## Helper functions

def piecewise_model_twosegments(x, b, m1, m2, x0):
    """
    Piecewise linear function with two segments

    Formulae:
        y1 = b + m1*x for segment 1 (x <= x0),
        y2 = y1(x0) + m2*(x-x0) for segment 2.

        y = y1 + y2.

    x : input
    b : intercept
    m1: slope 1
    m2: slope 2
    x0: joining
    """
    y = np.empty_like(x)

    mask1 = x <= x0
    mask2 = x > x0

    y0 = b + m1*x0
    y[mask1] = b + m1*x[mask1]
    y[mask2] = y0 + m2*(x[mask2] - x0)


    return y
    
    
```

```python
## Load given data
data = np.loadtxt("linearfit.dat")
x = data[:, 0].astype(float)
y = data[:, 1].astype(float)
```

```python
## initial guess of parameters

b_guess = y[len(x) // 2]
x0_guess = x[len(x) // 2] # midpoint
m1_guess = (y[len(x) // 2] - y[0])/(x0_guess - x[0])
m2_guess = (y[-1] - y[len(x) // 2])/(x[-1] - x0_guess)

#k = min(8, len(x))
#m1_guess = np.polyfit(x[:k], y[:k], 1)[0]
#b_guess  = np.mean(y[:k] - m1_guess*x[:k])
#dm_guess = 0.0  # start with no slope change
#x0_guess = x[len(x)//3]  # rough breakpoint guess

p0 = (b_guess, m1_guess, m2_guess, x0_guess)

# ---- bounds: keep x0 within data range ----
bounds = ([-np.inf, -np.inf, -np.inf, x.min()],
          [ np.inf,  np.inf,  np.inf, x.max()])

popt, pcov = curve_fit(piecewise_model_twosegments, x, y, p0=p0, bounds=bounds)
b, m1, m2, x0 = popt

print(f"Estimated breakpoint x0 = {x0:.6g}")
print(f"Linear-region slope (before x0): m1 = {m1:.6g}")
print(f"Post-break slope: m2 = {m1 + m2:.6g}")
```
