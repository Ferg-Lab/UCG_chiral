## Read AA chiral counts

```python
# all_temperatures = [2.0, 2.5, 2.7, 2.9, 3.0, 3.2, 3.5, 3.8, 4.0, 4.2, 4.4, 4.5, 4.6, 4.8, 5.0]
# all_temperatures = [2.0, 2.5, 2.7, 2.9, 3.0, 3.2, 3.5, 3.8, 4.0, 4.2, 4.3, 4.4, 4.5, 4.6, 4.8, 5.0]
# all_temperatures = [3.8, 4.2, 4.3, 4.5, 4.6, 5.0]
all_temperatures = [3.8, 4.2, 4.5, 4.6, 5.0]



```

```python
################################################################################################
# Read AA chiral count [See ../scripts/run_rdf.py  for calculation details]
################################################################################################
chiral_count_density = {}

for density in [0.07, 0.09, 0.11]:

    if density == 0.11:
        outpath = '/project2/andrewferguson/sivadasetty/doe/ucg/chiral_tetramer/SystemSize1000tetramers/'
    else:
        outpath = '/project2/andrewferguson/sivadasetty/doe/ucg/chiral_tetramer/SystemSize1000tetramers_'+ str(density) + '/'
    
    chiralcount_max_6 = {}

    if density == 0.11:
        for temperature in all_temperatures:
        
            aa_traj_file_path = outpath + '/T-'+str(temperature)+'_freqdouble_racemic/'
            
            chiral_results_file_max_6 = aa_traj_file_path +  'chiral_results/count_ld.pkl'
            
            with open(chiral_results_file_max_6, 'rb') as file:
                chiralcount_max_6[temperature] = pickle.load(file)

    else:
        for temperature in [3.8]:
        
            aa_traj_file_path = outpath + '/T-'+str(temperature)+'_freqdouble_racemic/'
            
            chiral_results_file_max_6 = aa_traj_file_path +  'chiral_results/count_ld.pkl'
            
            with open(chiral_results_file_max_6, 'rb') as file:
                chiralcount_max_6[temperature] = pickle.load(file)


    chiral_count_density[density] = chiralcount_max_6
```

```python
################################################################################################
# Read AA chiral count [See ../scripts/run_rdf.py  for calculation details]
################################################################################################
chiral_count_density_REM = {}

for density in [0.07, 0.09, 0.11]:

    if density == 0.11:
        outpath = '/project2/andrewferguson/sivadasetty/doe/ucg/chiral_tetramer/SystemSize1000tetramers/'
    else:
        outpath = '/project2/andrewferguson/sivadasetty/doe/ucg/chiral_tetramer/SystemSize1000tetramers_'+ str(density) + '/'
    
    chiralcount_max_6 = {}

    if density == 0.11:
        for temperature in all_temperatures:
        
            aa_traj_file_path = outpath + '/T-'+str(temperature)+'_freqdouble_racemic/'
            
            chiral_results_file_max_6 = aa_traj_file_path +  'chiral_results/count_ld_REM.pkl'
            
            with open(chiral_results_file_max_6, 'rb') as file:
                chiralcount_max_6[temperature] = pickle.load(file)

    else:
        for temperature in [3.8]:
        
            aa_traj_file_path = outpath + '/T-'+str(temperature)+'_freqdouble_racemic/'
            
            chiral_results_file_max_6 = aa_traj_file_path +  'chiral_results/count_ld_REM.pkl'
            
            with open(chiral_results_file_max_6, 'rb') as file:
                chiralcount_max_6[temperature] = pickle.load(file)


    chiral_count_density_REM[density] = chiralcount_max_6
```

## PLOT

```python
chiral_count_density_REM[0.11].keys()
```

```python
fig = plt.figure(figsize=(12, 8))
grid = GridSpec(3, 3)
plt.subplots_adjust(left=0.1, right=0.9, bottom=0.1, top=0.9, wspace=0.5, hspace=0.5)

j=0
k=0
axs = []

for l, density in enumerate(chiral_count_density_REM.keys()):

    print(density)

    for i, temperature in enumerate(chiral_count_density_REM[density].keys()):

        if k > 2:
            j += 1
            k -= 3

    
        ax11 = plt.subplot(grid[j, k])
        axs.append(ax11)
        ax11.grid()
        ax11.set_xlabel(r'Frames')
        ax11.set_ylabel(r'Count ($\rho$='+str(density)+', T='+str(temperature)+')')
    
        ax11.plot(list(chiral_count_density[density][temperature]['l'].keys()), 
                  list(chiral_count_density[density][temperature]['l'].values()), color='blue', label='L', alpha=0.25)
    
        ax11.twiny().scatter(list(chiral_count_density_REM[density][temperature]['l'].keys()),
                          list(chiral_count_density_REM[density][temperature]['l'].values()), color='blue', alpha=1, s=5)
    
        
        ax11.plot(list(chiral_count_density[density][temperature]['d'].keys()), 
                  list(chiral_count_density[density][temperature]['d'].values()), color='red', label='D', alpha=0.25)
    
        ax11.twiny().scatter(list(chiral_count_density_REM[density][temperature]['d'].keys()),
                          list(chiral_count_density_REM[density][temperature]['d'].values()), color='red', alpha=1, s=5)
    
        if i == 0 and l == 2:
            leg = ax11.legend(frameon=False)
            for lh in leg.legend_handles:
                lh.set_alpha(1) # Sets the alpha for the legend marker/handle

    
        k+= 1

 
# #####################################################################################
for i, ax in enumerate(axs):
    ax.text(-0.25, 1.18, '(' + string.ascii_lowercase[i] + ')', transform=ax.transAxes, size=15, weight='normal', va='top', ha='center')

    ax.yaxis.set_major_locator(MultipleLocator(200))
    ax.yaxis.set_minor_locator(AutoMinorLocator(n=10))

    if i >= 2:
        ax.xaxis.set_major_locator(MultipleLocator(4000))
    else:
        ax.xaxis.set_major_locator(MultipleLocator(1000))
    ax.xaxis.set_minor_locator(AutoMinorLocator(n=10))


```
